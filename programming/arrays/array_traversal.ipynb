<strong style = "font-size:30px; color:green">Array Traversal</strong>

```python
arr = [10, 20, 30, 40, 50, 60]

# 1. Count / length of elements
count = 0

for i in arr:
    count += 1

print("Count of elements :")
print(count)


# 2. Print elements in single line
print("Array elements :")
for i in arr:
    print(i, end=" ")
print()


# 3. Fetch all elements
print("Fetch all elements :")
for i in range(count):
    print(arr[i], end=" ")
print()


# 4. Fetch reverse order
print("Fetch elements in reverse order :")
for i in range(count - 1, -1, -1):
    print(arr[i], end=" ")
print()


# 5. Fetch first half forward
print("Fetch first half forward :")
for i in range(0, count // 2):
    print(arr[i], end=" ")
print()


# 6. Fetch first half backward
print("Fetch first half backward :")
for i in range(count // 2 - 1, -1, -1):
    print(arr[i], end=" ")
print()


# 7. Fetch second half forward
print("Fetch second half forward :")
for i in range(count // 2, count):
    print(arr[i], end=" ")
print()


# 8. Fetch second half backward
print("Fetch second half backward :")
for i in range(count - 1, count // 2 - 1, -1):
    print(arr[i], end=" ")
print()
```

<strong style="font-size:30px;color:green">Array Index based Traversal (even index)</strong>

```python
arr = [10, 20, 30, 40, 50, 60]

# Count
count = 0
for i in arr:
    count += 1


# 1. Fetch even indexed elements forward
print("Fetch even indexed elements forward :")
for i in range(0, count, 2):
    print(arr[i], end=" ")
print()


# 2. Fetch even indexed elements backward
print("Fetch even indexed elements backward :")
start = count - 1 if (count - 1) % 2 == 0 else count - 2
for i in range(start, -1, -2):
    print(arr[i], end=" ")
print()


# 3. Fetch first half even indexed forward
print("Fetch first half even indexed forward :")
for i in range(0, count // 2, 2):
    print(arr[i], end=" ")
print()


# 4. Fetch first half even indexed backward
print("Fetch first half even indexed backward :")
last = (count // 2) - 1
if last % 2 != 0:
    last -= 1

for i in range(last, -1, -2):
    print(arr[i], end=" ")
print()


# 5. Fetch second half even indexed forward
print("Fetch second half even indexed forward :")
start = count // 2
if start % 2 != 0:
    start += 1

for i in range(start, count, 2):
    print(arr[i], end=" ")
print()


# 6. Fetch second half even indexed backward
print("Fetch second half even indexed backward :")
last = count - 1
if last % 2 != 0:
    last -= 1

start = count // 2
if start % 2 != 0:
    start += 1

for i in range(last, start - 1, -2):
    print(arr[i], end=" ")
print()
```

<strong style="font-size:30px;color:green">Array Index based Traversal (odd index)</strong>

```python
arr = [10, 20, 30, 40, 50, 60]

# Count
count = 0
for i in arr:
    count += 1

# 1. Fetch only odd indexed elements forward
print("Fetch only odd indexed elements forward :")
for i in range(1, count, 2):
    print(arr[i], end=" ")
print()

# 2. Fetch only odd indexed elements reverse
print("Fetch only odd indexed elements reverse :")
start = count - 1 if (count - 1) % 2 == 1 else count - 2
for i in range(start, 0, -2):
    print(arr[i], end=" ")
print()

# 3. Fetch only first half odd indexed forward
print("Fetch only first half odd indexed forward :")
for i in range(1, count // 2, 2):
    print(arr[i], end=" ")
print()

# 4. Fetch only first half odd indexed reverse
print("Fetch only first half odd indexed reverse :")
last = (count // 2) - 1
if last % 2 == 0:
    last -= 1
for i in range(last, 0, -2):
    print(arr[i], end=" ")
print()

# 5. Fetch only second half odd indexed forward
print("Fetch only second half odd indexed forward :")
start = count // 2
if start % 2 == 0:
    start += 1
for i in range(start, count, 2):
    print(arr[i], end=" ")
print()

# 6. Fetch only second half odd indexed reversed
print("Fetch only second half odd indexed reversed :")
last = count - 1
if last % 2 == 0:
    last -= 1

start = count // 2
if start % 2 == 0:
    start += 1

for i in range(last, start - 1, -2):
    print(arr[i], end=" ")
print()
```

<strong style="font-size:30px;color:green">Count Even and Odd Indexed elements</strong>

```python
arr = [12, 7, 0, 25, 18, 31, 44, 9, 16, 5]

# Count
count = 0
for i in arr:
    count += 1


# 1. Count of even indexed elements
even_index_count = 0
for i in range(0, count, 2):
    even_index_count += 1

print("Count of even indexed elements :")
print(even_index_count)


# 2. Count of first half even indexed elements
first_half_even_index_count = 0
for i in range(0, count // 2, 2):
    first_half_even_index_count += 1

print("Count of first half even indexed elements :")
print(first_half_even_index_count)


# 3. Count of second half even indexed elements
second_half_even_index_count = 0

start = count // 2
if start % 2 != 0:
    start += 1

for i in range(start, count, 2):
    second_half_even_index_count += 1

print("Count of second half even indexed elements :")
print(second_half_even_index_count)


# 4. Count of odd indexed elements
odd_index_count = 0
for i in range(1, count, 2):
    odd_index_count += 1

print("Count of odd indexed elements :")
print(odd_index_count)


# 5. Count of first half odd indexed elements
first_half_odd_index_count = 0
for i in range(1, count // 2, 2):
    first_half_odd_index_count += 1

print("Count of first half odd indexed elements :")
print(first_half_odd_index_count)


# 6. Count of second half odd indexed elements
second_half_odd_index_count = 0

start = count // 2
if start % 2 == 0:
    start += 1

for i in range(start, count, 2):
    second_half_odd_index_count += 1

print("Count of second half odd indexed elements :")
print(second_half_odd_index_count)
```

<strong style="font-size:30px;color:green">Sum of even and odd indexed elements</strong>

```python
arr = [12, 7, 0, 25, 18, 31, 44, 9, 16, 5]

# Count
count = 0
for i in arr:
    count += 1


# 1. Sum of even indexed elements
even_index_sum = 0
for i in range(0, count, 2):
    even_index_sum += arr[i]

print("Sum of even indexed elements :")
print(even_index_sum)


# 2. Sum of first half even indexed elements
first_half_even_index_sum = 0
for i in range(0, count // 2, 2):
    first_half_even_index_sum += arr[i]

print("Sum of first half even indexed elements :")
print(first_half_even_index_sum)


# 3. Sum of second half even indexed elements
second_half_even_index_sum = 0

start = count // 2
if start % 2 != 0:
    start += 1

for i in range(start, count, 2):
    second_half_even_index_sum += arr[i]

print("Sum of second half even indexed elements :")
print(second_half_even_index_sum)


# 4. Sum of odd indexed elements
odd_index_sum = 0
for i in range(1, count, 2):
    odd_index_sum += arr[i]

print("Sum of odd indexed elements :")
print(odd_index_sum)


# 5. Sum of first half odd indexed elements
first_half_odd_index_sum = 0
for i in range(1, count // 2, 2):
    first_half_odd_index_sum += arr[i]

print("Sum of first half odd indexed elements :")
print(first_half_odd_index_sum)


# 6. Sum of second half odd indexed elements
second_half_odd_index_sum = 0

start = count // 2
if start % 2 == 0:
    start += 1

for i in range(start, count, 2):
    second_half_odd_index_sum += arr[i]

print("Sum of second half odd indexed elements :")
print(second_half_odd_index_sum)
```
