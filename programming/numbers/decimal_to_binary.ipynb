<strong style="font-size: 30px; color: green;">Without using in-built</strong>

```python
num = int(input("Enter a decimal number: "))
print("You Entered",num)
if num == 0:
    print("Binary: 0")
else:
    binary = ""
    while num > 0:
        binary = str(num % 2) + binary
        num //= 2
    print("Binary value is :", binary)
```

<strong style="font-size: 30px; color: green;">Using recursion</strong>

```python
def decimal_to_binary(n):
    if n <= 1:
        return str(n)
    return decimal_to_binary(n // 2) + str(n % 2)

num = int(input("Enter a decimal number: "))
print("You Entered",num)
print("Binary value is :", decimal_to_binary(num))
```
