<strong style="font-size:30px;color:green">Remove duplicate words from a sentence while preserving their original order</strong>

<strong style="font-size:25px;color:green">Using dict.fromkeys()</strong>

```python
sentence = "Python is fun and Python is powerful and learning Python is rewarding"
print("You Entered :", sentence)

words = sentence.split()
unique_words = list(dict.fromkeys(words))
result = " ".join(unique_words)

print("Result Sentence :", result)
```

<strong style="font-size:25px;color:green">Using a set for tracking and a list to preserve order</strong>

```python
sentence = "Python is fun and Python is powerful and learning Python is rewarding"
print("You Entered :", sentence)

words = sentence.split()
seen = set()
unique_words = []

for word in words:
    if word not in seen:
        seen.add(word)
        unique_words.append(word)

result = " ".join(unique_words)
print("Result Sentence :", result)
```

<strong style="font-size:25px;color:green">Using collections.OrderedDict</strong>

```python
from collections import OrderedDict

sentence = "Python is fun and Python is powerful and learning Python is rewarding"
print("You Entered :", sentence)

words = sentence.split()
unique_words = list(OrderedDict.fromkeys(words).keys())
result = " ".join(unique_words)

print("Result Sentence :", result)
```

<strong style="font-size:25px;color:green">Without using split() - manual word extraction and set</strong>

```python
sentence = "Python is fun and Python is powerful and learning Python is rewarding"
print("You Entered :", sentence)

# Extract words manually without split()
words = []
current_word = ""
for char in sentence:
    if char != " ":
        current_word += char
    else:
        if current_word:
            words.append(current_word)
            current_word = ""
if current_word:
    words.append(current_word)

seen = set()
unique_words = []
for word in words:
    if word not in seen:
        seen.add(word)
        unique_words.append(word)

# Reconstruct sentence manually without join()
result = ""
for i in range(len(unique_words)):
    if i > 0:
        result += " "
    result += unique_words[i]

print("Result Sentence :", result)
```

<strong style="font-size:25px;color:green">Without using built-in functions, methods, or sets</strong>

```python
sentence = "Python is fun and Python is powerful and learning Python is rewarding"
print("You Entered :", sentence)

# Extract words manually without split()
words = []
current_word = ""
for char in sentence:
    if char != " ":
        current_word += char
    else:
        if current_word != "":
            words.append(current_word)
            current_word = ""
if current_word != "":
    words.append(current_word)

# Remove duplicates while preserving order using linear scan
unique_words = []
for word in words:
    is_duplicate = False
    for item in unique_words:
        if item == word:
            is_duplicate = True
            break
    if not is_duplicate:
        unique_words.append(word)

# Reconstruct sentence without join()
result = ""
for i in range(len(unique_words)):
    if i > 0:
        result += " "
    result += unique_words[i]

print("Result Sentence :", result)
```
