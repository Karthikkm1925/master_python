<strong style="font-size:30px;color:green">Remove all occurrences of a substring without using replace()</strong>

<strong style="font-size:25px;color:green">Using split() and join()</strong>

```python
main_str = "banana"
sub_str = "an"
print("Main String :", main_str)
print("Substring   :", sub_str)

result = "".join(main_str.split(sub_str))

print("Result :", result)
```

<strong style="font-size:25px;color:green">Using a while loop and string slicing</strong>

```python
main_str = "banana"
sub_str = "an"
print("Main String :", main_str)
print("Substring   :", sub_str)

result = ""
main_len = len(main_str)
sub_len = len(sub_str)
i = 0

while i < main_len:
    if main_str[i : i + sub_len] == sub_str:
        i += sub_len
    else:
        result += main_str[i]
        i += 1

print("Result :", result)
```

<strong style="font-size:25px;color:green">Using str.find() and list accumulation</strong>

```python
main_str = "banana"
sub_str = "an"
print("Main String :", main_str)
print("Substring   :", sub_str)

result = []
sub_len = len(sub_str)
start = 0

while True:
    pos = main_str.find(sub_str, start)
    if pos == -1:
        result.append(main_str[start:])
        break
    result.append(main_str[start:pos])
    start = pos + sub_len

result = "".join(result)

print("Result :", result)
```

<strong style="font-size:25px;color:green">Using regular expressions (re.sub)</strong>

```python
import re

main_str = "banana"
sub_str = "an"
print("Main String :", main_str)
print("Substring   :", sub_str)

result = re.sub(re.escape(sub_str), "", main_str)

print("Result :", result)
```

<strong style="font-size:25px;color:green">Without using built-in methods or libraries (Nested while loops)</strong>

```python
main_str = "banana"
sub_str = "an"
print("Main String :", main_str)
print("Substring   :", sub_str)

result = ""
main_len = len(main_str)
sub_len = len(sub_str)
i = 0

while i < main_len:
    match = True
    if i <= main_len - sub_len:
        j = 0
        while j < sub_len:
            if main_str[i + j] != sub_str[j]:
                match = False
                break
            j += 1
    else:
        match = False

    if match:
        i += sub_len
    else:
        result += main_str[i]
        i += 1

print("Result :", result)
```
