<strong style="font-size:30px;color:green">Rreverse a string in Python without using any built-in/string reverse function”</strong>

<strong style="font-size:25px;color:green">Using a for loop</strong>

```python
s = "python"

reverse = ""

for i in range(len(s) - 1, -1, -1):
    reverse = reverse + s[i]
print("Before :",s)
print("After :",reverse)
```

<strong style="font-size:25px;color:green">Using a while loop</strong>

```python
s = "python"

reverse = ""
i = len(s) - 1

while i >= 0:
    reverse = reverse + s[i]
    i = i - 1
print("Before :",s)
print("After :",reverse)
```

<strong style="font-size:25px;color:green">Using a for loop from 0 and prepend</strong>

```python
s = "python"

reverse = ""

for ch in s:
    reverse = ch + reverse

print("Before :",s)
print("After :",reverse)
```

<strong style="font-size:25px;color:green">Using slicing</strong>

```python
s = "python"

reverse = s[::-1]

print("Before :",s)
print("After :",reverse)
```

<strong style="font-size:25px;color:green">Using reversed()</strong>

```python
s = "python"

reverse = ""

for ch in reversed(s):
    reverse = reverse + ch
print("Before :",s)
print("After :",reverse)
```

<strong style="font-size:25px;color:green">Using recursion</strong>

```python
def reverse_string(s):
    if len(s) == 0:
        return ""

    return reverse_string(s[1:]) + s[0]


s = "python"
print("Before :",s)
print("After :",reverse_string(s))
```

<strong style="font-size:25px;color:green">Recursion without slicing</strong>

```python
def reverse_string(s, index):
    if index < 0:
        return ""

    return s[index] + reverse_string(s, index - 1)


s = "python"
print("Before :",s)
print("After :",reverse_string(s, len(s) - 1))
```

<strong style="font-size:25px;color:green">Using a list manually</strong>

```python
s = "python"

chars = []

for ch in s:
    chars.append(ch)

reverse = ""

i = len(chars) - 1

while i >= 0:
    reverse = reverse + chars[i]
    i -= 1

print("Before :",s)
print("After :",reverse)
```

<strong style="font-size:25px;color:green">Using two-pointer swapping</strong>

```python
s = "python"

chars = []

for ch in s:
    chars.append(ch)

left = 0
right = len(chars) - 1

while left < right:
    temp = chars[left]
    chars[left] = chars[right]
    chars[right] = temp

    left += 1
    right -= 1

reverse = ""

for ch in chars:
    reverse = reverse + ch

print("Before :",s)
print("After :",reverse)
```

<strong style="font-size:25px;color:green">Using a stack concept</strong>

```python
s = "python"

stack = []

for ch in s:
    stack.append(ch)

reverse = ""

while len(stack) > 0:
    reverse = reverse + stack[len(stack) - 1]
    stack.pop()
print("Before :",s)
print("After :",reverse)
```

<strong style="font-size:25px;color:green">Using range() with a positive step and index calculation</strong>

```python
s = "python"

reverse = ""

n = len(s)

for i in range(n):
    reverse = reverse + s[n - 1 - i]

print("Before :",s)
print("After :",reverse)
```

<strong style="font-size:25px;color:green">Using a while loop</strong>

```python
s = "python"

n = len(s)

result = [""] * n

for i in range(n):
    result[n - 1 - i] = s[i]

reverse = ""

for ch in result:
    reverse += ch
print("Before :",s)
print("After :",reverse)
```

<strong style="font-size:25px;color:green">Using character position</strong>

```python
s = "python"

reverse = ""

i = len(s)

while i > 0:
    i -= 1
    reverse += s[i]

print("Before :",s)
print("After :",reverse)
```

<strong style="font-size:25px;color:green">Using recursion with an accumulator</strong>

```python
def reverse_string(s, index, result):
    if index < 0:
        return result

    result += s[index]

    return reverse_string(s, index - 1, result)


s = "python"
print("Before :",s)
print("After",reverse_string(s, len(s) - 1, ""))
```

<strong style="font-size:25px;color:green">Using a custom function with two pointers</strong>

```python
def reverse_string(s):

    chars = []

    for ch in s:
        chars.append(ch)

    left = 0
    right = len(chars) - 1

    while left < right:

        temp = chars[left]
        chars[left] = chars[right]
        chars[right] = temp

        left += 1
        right -= 1

    result = ""

    for ch in chars:
        result += ch

    return result

print("Before :",s)
print("After :",reverse_string("python"))
```

<strong style="font-size:25px;color:green">Character-by-character recursion</strong>

```python
def reverse_string(s, index=0):

    if index == len(s):
        return ""

    return reverse_string(s, index + 1) + s[index]

print("Before : python")
print("After :",reverse_string("python"))
```
