<strong style="font-size:30px;color:green">Find the last occurrence of a specific character in a string</strong>

<strong style="font-size:25px;color:green">Using rfind() method</strong>

```python
s = "programming"
target = "r"
print("You Entered :", s)
print("Target Character :", target)

last_index = s.rfind(target)

if last_index != -1:
    print("Last occurrence index:", last_index)
else:
    print("Character not found.")
```

<strong style="font-size:25px;color:green">Using reverse loop</strong>

```python
s = "programming"
target = "r"
print("You Entered :", s)
print("Target Character :", target)

last_index = -1
for i in range(len(s) - 1, -1, -1):
    if s[i] == target:
        last_index = i
        break

if last_index != -1:
    print("Last occurrence index:", last_index)
else:
    print("Character not found.")
```

<strong style="font-size:25px;color:green">Using forward loop and index update</strong>

```python
s = "programming"
target = "r"
print("You Entered :", s)
print("Target Character :", target)

last_index = -1
for i in range(len(s)):
    if s[i] == target:
        last_index = i

if last_index != -1:
    print("Last occurrence index:", last_index)
else:
    print("Character not found.")
```

<strong style="font-size:25px;color:green">Using rindex() method</strong>

```python
s = "programming"
target = "r"
print("You Entered :", s)
print("Target Character :", target)

try:
    last_index = s.rindex(target)
    print("Last occurrence index:", last_index)
except ValueError:
    print("Character not found.")
```

<strong style="font-size:25px;color:green">Using recursion</strong>

```python
def find_last_occurrence(s, target, index=None):
    if index is None:
        index = len(s) - 1
    if index < 0:
        return -1
    if s[index] == target:
        return index
    return find_last_occurrence(s, target, index - 1)


s = "programming"
target = "r"
print("You Entered :", s)
print("Target Character :", target)

last_index = find_last_occurrence(s, target)

if last_index != -1:
    print("Last occurrence index:", last_index)
else:
    print("Character not found.")
```
