<strong style="font-size:30px;color:green">Check if a string is a valid email address</strong>

<strong style="font-size:25px;color:green">Using basic loops</strong>

```python
email = "user@gmail.com"
print("You Entered :", email)

at_count = 0
dot_count = 0
at_position = -1
dot_position = -1

for i in range(len(email)):
    if email[i] == "@":
        at_count += 1
        at_position = i

    if email[i] == ".":
        dot_count += 1
        dot_position = i

if at_count == 1 and dot_count >= 1:
    if at_position > 0 and at_position < len(email) - 1:
        if dot_position > at_position + 1 and dot_position < len(email) - 1:
            print("Valid email")
        else:
            print("Invalid email")
    else:
        print("Invalid email")
else:
    print("Invalid email")
```

<strong style="font-size:25px;color:green">Checking username and domain separately</strong>

```python
email = "user@gmail.com"
print("You Entered :", email)

at_position = -1

for i in range(len(email)):
    if email[i] == "@":
        at_position = i
        break

if at_position > 0 and at_position < len(email) - 1:

    username = ""
    domain = ""

    i = 0

    while i < at_position:
        username += email[i]
        i += 1

    i = at_position + 1

    while i < len(email):
        domain += email[i]
        i += 1

    dot_found = False

    for ch in domain:
        if ch == ".":
            dot_found = True
            break

    if dot_found and len(username) > 0 and len(domain) > 2:
        print("Valid email")
    else:
        print("Invalid email")

else:
    print("Invalid email")
```

<strong style="font-size:25px;color:green">Checking every character</strong>

```python
email = "user@gmail.com"
print("You Entered :", email)

valid = True
at_count = 0
dot_after_at = False

for i in range(len(email)):

    ch = email[i]

    if ch == "@":
        at_count += 1

        if i == 0 or i == len(email) - 1:
            valid = False

    elif ch == ".":
        if i == 0 or i == len(email) - 1:
            valid = False

        if at_count == 1:
            dot_after_at = True

if at_count != 1:
    valid = False

if not dot_after_at:
    valid = False

if valid:
    print("Valid email")
else:
    print("Invalid email")
```

<strong style="font-size:25px;color:green">Using regular expression</strong>

```python
import re

email = "user@gmail.com"
print("You Entered :", email)

pattern = r"^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$"

if re.match(pattern, email):
    print("Valid email")
else:
    print("Invalid email")
```
