<strong style="font-size:30px;color:green">Print the characters at even and odd indexes separately</strong>

<strong style="font-size:25px;color:green">Using a for loop and range()</strong>

```python
s = "Python"
print("You Entered :", s)

print("Even index characters:")
for i in range(0, len(s), 2):
    print(s[i], end=" ")

print("\nOdd index characters:")
for i in range(1, len(s), 2):
    print(s[i], end=" ")
print()
```

<strong style="font-size:25px;color:green">Using a for loop and the modulus operator</strong>

```python
s = "Python"
print("You Entered :", s)

print("Even index characters:")
for i in range(len(s)):
    if i % 2 == 0:
        print(s[i], end=" ")

print("\nOdd index characters:")
for i in range(len(s)):
    if i % 2 != 0:
        print(s[i], end=" ")
print()
```

<strong style="font-size:25px;color:green">Using a while loop</strong>

```python
s = "Python"
print("You Entered :", s)

i = 0
print("Even index characters:")
while i < len(s):
    print(s[i], end=" ")
    i += 2

i = 1
print("\nOdd index characters:")
while i < len(s):
    print(s[i], end=" ")
    i += 2
print()
```

<strong style="font-size:25px;color:green">Using a single for loop</strong>

```python
s = "Python"
print("You Entered :", s)

for i in range(len(s)):
    if i % 2 == 0:
        print("Even:", s[i])
    else:
        print("Odd:", s[i])
```

<strong style="font-size:25px;color:green">Using string slicing</strong>

```python
s = "Python"
print("You Entered :", s)

print("Even index characters:", s[::2])
print("Odd index characters:", s[1::2])
```

<strong style="font-size:25px;color:green">Using manual index tracking with a for loop</strong>

```python
s = "Python"
print("You Entered :", s)

i = 0
for ch in s:
    if i % 2 == 0:
        print("Even index:", ch)
    else:
        print("Odd index:", ch)
    i += 1
```
