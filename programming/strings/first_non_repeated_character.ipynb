<strong style="font-size:30px;color:green">Find the first non-repeated character in a string</strong>

<strong style="font-size:25px;color:green">Using nested loops</strong>

```python
s = "swiss"
print("You Entered :", s)

first_non_repeated = None

for i in range(len(s)):
    is_repeated = False
    for j in range(len(s)):
        if i != j and s[i] == s[j]:
            is_repeated = True
            break
    if not is_repeated:
        first_non_repeated = s[i]
        break

if first_non_repeated:
    print("First non-repeated character:", first_non_repeated)
else:
    print("No non-repeated character found.")
```

<strong style="font-size:25px;color:green">Using a frequency dictionary</strong>

```python
s = "swiss"
print("You Entered :", s)

freq = {}

for ch in s:
    if ch in freq:
        freq[ch] += 1
    else:
        freq[ch] = 1

first_non_repeated = None

for ch in s:
    if freq[ch] == 1:
        first_non_repeated = ch
        break

if first_non_repeated:
    print("First non-repeated character:", first_non_repeated)
else:
    print("No non-repeated character found.")
```

<strong style="font-size:25px;color:green">Using string count() method</strong>

```python
s = "swiss"
print("You Entered :", s)

first_non_repeated = None

for ch in s:
    if s.count(ch) == 1:
        first_non_repeated = ch
        break

if first_non_repeated:
    print("First non-repeated character:", first_non_repeated)
else:
    print("No non-repeated character found.")
```

<strong style="font-size:25px;color:green">Using collections.Counter</strong>

```python
from collections import Counter

s = "swiss"
print("You Entered :", s)

counts = Counter(s)
first_non_repeated = None

for ch in s:
    if counts[ch] == 1:
        first_non_repeated = ch
        break

if first_non_repeated:
    print("First non-repeated character:", first_non_repeated)
else:
    print("No non-repeated character found.")
```

<strong style="font-size:25px;color:green">Without built-in functions or dictionary</strong>

```python
s = "swiss"
print("You Entered :", s)

first_non_repeated = None

for i in range(len(s)):
    count = 0
    for j in range(len(s)):
        if s[i] == s[j]:
            count += 1
    if count == 1:
        first_non_repeated = s[i]
        break

if first_non_repeated:
    print("First non-repeated character:", first_non_repeated)
else:
    print("No non-repeated character found.")
```
