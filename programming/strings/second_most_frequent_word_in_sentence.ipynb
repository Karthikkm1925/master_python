<strong style="font-size:30px;color:green">Find the second most frequent word in a sentence</strong>

<strong style="font-size:25px;color:green">Using collections.Counter and most_common()</strong>

```python
from collections import Counter

sentence = "apple orange banana apple apple orange banana apple grape apple orange"
print("You Entered :", sentence)

words = sentence.split()
counts = Counter(words)

# Sort words by frequency descending
sorted_words = counts.most_common()
max_freq = sorted_words[0][1]

second_word = None
second_freq = 0
for word, count in sorted_words:
    if count < max_freq:
        second_word = word
        second_freq = count
        break

print("Second Most Frequent Word :", second_word)
print("Frequency :", second_freq)
```

<strong style="font-size:25px;color:green">Using a frequency dictionary and sorting</strong>

```python
sentence = "apple orange banana apple apple orange banana apple grape apple orange"
print("You Entered :", sentence)

words = sentence.split()
freq = {}
for word in words:
    freq[word] = freq.get(word, 0) + 1

sorted_items = sorted(freq.items(), key=lambda x: x[1], reverse=True)
max_freq = sorted_items[0][1]

second_word = None
second_freq = 0
for word, count in sorted_items:
    if count < max_freq:
        second_word = word
        second_freq = count
        break

print("Second Most Frequent Word :", second_word)
print("Frequency :", second_freq)
```

<strong style="font-size:25px;color:green">Tracking top two frequencies in a single pass</strong>

```python
sentence = "apple orange banana apple apple orange banana apple grape apple orange"
print("You Entered :", sentence)

words = sentence.split()
freq = {}
for word in words:
    freq[word] = freq.get(word, 0) + 1

first_word, second_word = None, None
first_max, second_max = -1, -1

for word, count in freq.items():
    if count > first_max:
        second_max = first_max
        second_word = first_word
        first_max = count
        first_word = word
    elif count < first_max and count > second_max:
        second_max = count
        second_word = word

print("Second Most Frequent Word :", second_word)
print("Frequency :", second_max)
```

<strong style="font-size:25px;color:green">Without using split() - manual word extraction and dictionary</strong>

```python
sentence = "apple orange banana apple apple orange banana apple grape apple orange"
print("You Entered :", sentence)

# Extract words manually without split()
words = []
current_word = ""
for char in sentence:
    if char != " ":
        current_word += char
    else:
        if current_word:
            words.append(current_word)
            current_word = ""
if current_word:
    words.append(current_word)

freq = {}
for word in words:
    if word in freq:
        freq[word] += 1
    else:
        freq[word] = 1

max_freq = -1
for count in freq.values():
    if count > max_freq:
        max_freq = count

second_max = -1
second_word = None
for word, count in freq.items():
    if second_max < count < max_freq:
        second_max = count
        second_word = word

print("Second Most Frequent Word :", second_word)
print("Frequency :", second_max)
```

```python
sentence = "apple orange banana apple apple orange banana apple grape apple orange"
print("You Entered :", sentence)

# Extract words manually
words = []
current_word = ""
for char in sentence:
    if char != " ":
        current_word += char
    else:
        if current_word != "":
            words.append(current_word)
            current_word = ""
if current_word != "":
    words.append(current_word)

# Track unique words and counts using parallel lists
unique_words = []
counts = []

for word in words:
    found = False
    for i in range(len(unique_words)):
        if unique_words[i] == word:
            counts[i] += 1
            found = True
            break
    if not found:
        unique_words.append(word)
        counts.append(1)

# Find first maximum frequency
first_max = -1
for c in counts:
    if c > first_max:
        first_max = c

# Find second maximum frequency
second_max = -1
second_word = None
for i in range(len(counts)):
    if counts[i] < first_max and counts[i] > second_max:
        second_max = counts[i]
        second_word = unique_words[i]

print("Second Most Frequent Word :", second_word)
print("Frequency :", second_max)
```
