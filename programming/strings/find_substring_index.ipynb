<strong style="font-size:30px;color:green">Find the index of the first occurrence of a substring in a string</strong>

<strong style="font-size:25px;color:green">Using string slicing in a for loop</strong>

```python
main_str = "hello world"
sub_str = "world"
print("Main String :", main_str)
print("Substring   :", sub_str)

index = -1
main_len = len(main_str)
sub_len = len(sub_str)

for i in range(main_len - sub_len + 1):
    if main_str[i : i + sub_len] == sub_str:
        index = i
        break

print("First Occurrence Index :", index)
```

<strong style="font-size:25px;color:green">Using nested loops (Character-by-character check)</strong>

```python
main_str = "python programming"
sub_str = "java"
print("Main String :", main_str)
print("Substring   :", sub_str)

index = -1
main_len = len(main_str)
sub_len = len(sub_str)

for i in range(main_len - sub_len + 1):
    match = True
    for j in range(sub_len):
        if main_str[i + j] != sub_str[j]:
            match = False
            break
    if match == True:
        index = i
        break

print("First Occurrence Index :", index)
```

<strong style="font-size:25px;color:green">Using a while loop</strong>

```python
main_str = "banana"
sub_str = "nan"
print("Main String :", main_str)
print("Substring   :", sub_str)

index = -1
main_len = len(main_str)
sub_len = len(sub_str)
i = 0

while i <= main_len - sub_len:
    match = True
    j = 0
    while j < sub_len:
        if main_str[i + j] != sub_str[j]:
            match = False
            break
        j = j + 1

    if match == True:
        index = i
        break

    i = i + 1

print("First Occurrence Index :", index)
```

<strong style="font-size:25px;color:green">Using a custom helper function</strong>

```python
def check_match(main_str, sub_str, start_index):
    for j in range(len(sub_str)):
        if main_str[start_index + j] != sub_str[j]:
            return False
    return True

main_str = "coding interview"
sub_str = "test"
print("Main String :", main_str)
print("Substring   :", sub_str)

index = -1
main_len = len(main_str)
sub_len = len(sub_str)

for i in range(main_len - sub_len + 1):
    if check_match(main_str, sub_str, i):
        index = i
        break

print("First Occurrence Index :", index)
```

<strong style="font-size:25px;color:green">Using recursion</strong>

```python
def find_index_recursive(main_str, sub_str, i):
    main_len = len(main_str)
    sub_len = len(sub_str)

    if i > main_len - sub_len:
        return -1

    match = True
    for j in range(sub_len):
        if main_str[i + j] != sub_str[j]:
            match = False
            break

    if match == True:
        return i

    return find_index_recursive(main_str, sub_str, i + 1)

main_str = "abcdefg"
sub_str = "cde"
print("Main String :", main_str)
print("Substring   :", sub_str)

result = find_index_recursive(main_str, sub_str, 0)
print("First Occurrence Index :", result)
```
