<strong style="font-size:30px;color:green">Count alphabets, digits, spaces, and special characters in a string</strong>

<strong style="font-size:25px;color:green">Using built-in string methods</strong>

```python
s = "Python 3.11 @ #1"
print("You Entered :", s)

alphabets = 0
digits = 0
spaces = 0
special = 0

for ch in s:
    if ch.isalpha():
        alphabets += 1
    elif ch.isdigit():
        digits += 1
    elif ch.isspace():
        spaces += 1
    else:
        special += 1

print("Alphabets:", alphabets)
print("Digits:", digits)
print("Spaces:", spaces)
print("Special Characters:", special)
```

<strong style="font-size:25px;color:green">Using ASCII character range comparison</strong>

```python
s = "Python 3.11 @ #1"
print("You Entered :", s)

alphabets = 0
digits = 0
spaces = 0
special = 0

for ch in s:
    if ("a" <= ch <= "z") or ("A" <= ch <= "Z"):
        alphabets += 1
    elif "0" <= ch <= "9":
        digits += 1
    elif ch == " ":
        spaces += 1
    else:
        special += 1

print("Alphabets:", alphabets)
print("Digits:", digits)
print("Spaces:", spaces)
print("Special Characters:", special)
```

<strong style="font-size:25px;color:green">Using sum() and generator expressions</strong>

```python
s = "Python 3.11 @ #1"
print("You Entered :", s)

alphabets = sum(1 for ch in s if ch.isalpha())
digits = sum(1 for ch in s if ch.isdigit())
spaces = sum(1 for ch in s if ch.isspace())
special = sum(1 for ch in s if not (ch.isalpha() or ch.isdigit() or ch.isspace()))

print("Alphabets:", alphabets)
print("Digits:", digits)
print("Spaces:", spaces)
print("Special Characters:", special)
```

<strong style="font-size:25px;color:green">Using a dictionary counter</strong>

```python
s = "Python 3.11 @ #1"
print("You Entered :", s)

counts = {"alphabets": 0, "digits": 0, "spaces": 0, "special": 0}

for ch in s:
    if ch.isalpha():
        counts["alphabets"] += 1
    elif ch.isdigit():
        counts["digits"] += 1
    elif ch.isspace():
        counts["spaces"] += 1
    else:
        counts["special"] += 1

print("Alphabets:", counts["alphabets"])
print("Digits:", counts["digits"])
print("Spaces:", counts["spaces"])
print("Special Characters:", counts["special"])
```

<strong style="font-size:25px;color:green">Using regular expressions</strong>

```python
import re

s = "Python 3.11 @ #1"
print("You Entered :", s)

alphabets = len(re.findall(r"[a-zA-Z]", s))
digits = len(re.findall(r"[0-9]", s))
spaces = len(re.findall(r"\s", s))
special = len(re.findall(r"[^a-zA-Z0-9\s]", s))

print("Alphabets:", alphabets)
print("Digits:", digits)
print("Spaces:", spaces)
print("Special Characters:", special)
```
