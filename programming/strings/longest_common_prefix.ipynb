<strong style="font-size:30px;color:green">Find the longest common prefix among a list of strings</strong>

<strong style="font-size:25px;color:green">Vertical scanning</strong>

```python
strs = ["flower", "flow", "flight"]
print("You Entered :", strs)

prefix = ""

if strs:
    for i in range(len(strs[0])):
        ch = strs[0][i]
        for s in strs[1:]:
            if i >= len(s) or s[i] != ch:
                ch = ""
                break
        if ch == "":
            break
        prefix += ch

print("Longest common prefix:", prefix)
```

<strong style="font-size:25px;color:green">Horizontal scanning</strong>

```python
strs = ["flower", "flow", "flight"]
print("You Entered :", strs)

if not strs:
    prefix = ""
else:
    prefix = strs[0]
    for i in range(1, len(strs)):
        while not strs[i].startswith(prefix):
            prefix = prefix[:-1]
            if not prefix:
                break

print("Longest common prefix:", prefix)
```

<strong style="font-size:25px;color:green">Using sorting</strong>

```python
strs = ["flower", "flow", "flight"]
print("You Entered :", strs)

if not strs:
    prefix = ""
else:
    sorted_strs = sorted(strs)
    first = sorted_strs[0]
    last = sorted_strs[-1]
    prefix = ""
    for i in range(min(len(first), len(last))):
        if first[i] == last[i]:
            prefix += first[i]
        else:
            break

print("Longest common prefix:", prefix)
```

<strong style="font-size:25px;color:green">Using zip() and set()</strong>

```python
strs = ["flower", "flow", "flight"]
print("You Entered :", strs)

prefix = ""
for chars in zip(*strs):
    if len(set(chars)) == 1:
        prefix += chars[0]
    else:
        break

print("Longest common prefix:", prefix)
```

<strong style="font-size:25px;color:green">Without built-in functions</strong>

```python
strs = ["flower", "flow", "flight"]
print("You Entered :", strs)

if not strs:
    prefix = ""
else:
    min_len = len(strs[0])
    for s in strs:
        count = 0
        for ch in s:
            count += 1
        if count < min_len:
            min_len = count

    prefix = ""
    for i in range(min_len):
        current_ch = strs[0][i]
        match = True
        for j in range(1, len(strs)):
            if strs[j][i] != current_ch:
                match = False
                break
        if match:
            prefix += current_ch
        else:
            break

print("Longest common prefix:", prefix)
```
