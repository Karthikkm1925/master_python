<strong style="font-size:30px;color:green">Remove all duplicates from a string</strong>

<strong style="font-size:25px;color:green">Using nested loops</strong>

```python
string = "programming"
print("You Entered :",string)
result = ""

for i in range(len(string)):
    duplicate = False

    for j in range(len(result)):
        if string[i] == result[j]:
            duplicate = True
            break

    if duplicate == False:
        result = result + string[i]

print("After removing duplicates :",result)
```

<strong style="font-size:25px;color:green">Using in</strong>

```python
string = "programming"
print("You Entered :",string)
result = ""

for char in string:
    if char not in result:
        result = result + char

print("After removing duplicates :",result)
```

<strong style="font-size:25px;color:green">Using a while loop</strong>

```python
string = "programming"
print("You Entered",string)
result = ""

i = 0

while i < len(string):

    duplicate = False
    j = 0

    while j < len(result):

        if string[i] == result[j]:
            duplicate = True
            break

        j = j + 1

    if duplicate == False:
        result = result + string[i]

    i = i + 1

print("After removing duplicates :",result)
```

<strong style="font-size:25px;color:green">Using an index comparison</strong>

```python
string = "programming"
result = ""
print("You Entered :",string)
for i in range(len(string)):

    found = False

    for j in range(i):

        if string[i] == string[j]:
            found = True
            break

    if found == False:
        result = result + string[i]

print("After removing duplicates :",result)
```

<strong style="font-size:25px;color:green">Using a list manually</strong>

```python
string = "programming"
print("You Entered",string)
result = []

for char in string:

    found = False

    for item in result:

        if char == item:
            found = True
            break

    if found == False:
        result.append(char)

final_result = ""

for char in result:
    final_result = final_result + char

print("After removing duplicates :",final_result)
```

<strong style="font-size:25px;color:green">Using recursion</strong>

```python
def remove_duplicates(string, index, result):

    if index == len(string):
        return result

    found = False

    for char in result:
        if string[index] == char:
            found = True
            break

    if found == False:
        result = result + string[index]

    return remove_duplicates(string, index + 1, result)


string = "programming"
print("You Entered :",string)
result = remove_duplicates(string, 0, "")

print("After removing duplicates :",result)
```
