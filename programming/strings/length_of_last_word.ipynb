<strong style="font-size:30px;color:green">Find the length of the last word in a sentence</strong>

<strong style="font-size:25px;color:green">Traverse from the end</strong>

```python
sentence = input("Enter a sentence: ")
print("You Entered :",sentence)
count = 0
i = 0

while sentence[i:]:
    i += 1

# Move backwards over trailing spaces
i -= 1

while i >= 0 and sentence[i] == ' ':
    i -= 1

# Count the last word
while i >= 0 and sentence[i] != ' ':
    count += 1
    i -= 1

print("Length of last word:", count)
```

<strong style="font-size:25px;color:green">Using for loop</strong>

```python
sentence = input("Enter a sentence: ")
print("You Entered :",sentence)
count = 0
current = 0

for ch in sentence:
    if ch == ' ':
        current = 0
    else:
        current += 1
        count = current

print("Length of last word:", count)
```

<strong style="font-size:25px;color:green">Using while loop</strong>

```python
sentence = input("Enter a sentence: ")
print("You Entered :",sentence)
i = 0

while True:
    try:
        ch = sentence[i]
    except IndexError:
        break

    i += 1

count = 0

i -= 1

while i >= 0 and sentence[i] == ' ':
    i -= 1

while i >= 0 and sentence[i] != ' ':
    count += 1
    i -= 1

print("Length of last word:", count)
```

<strong style="font-size:25px;color:green">Without len(), split(), strip() — using for</strong>

```python
sentence = input("Enter a sentence: ")
print("You Entered :",sentence)
last_word_length = 0

for character in sentence:
    if character == ' ':
        last_word_length = 0
    else:
        last_word_length += 1

print("Length of last word:", last_word_length)
```

<strong style="font-size:25px;color:green">Handle multiple spaces at the end</strong>

```python
sentence = input("Enter a sentence: ")
print("You Entered :",sentence)
current = 0
last_word_length = 0

for character in sentence:
    if character == ' ':
        if current != 0:
            last_word_length = current
            current = 0
    else:
        current += 1

if current != 0:
    last_word_length = current

print("Length of last word:", last_word_length)
```

<strong style="font-size:25px;color:green">Without any built-in function/method — manual character counting</strong>

```python
sentence = input("Enter a sentence: ")
print("You Entered :",sentence)
last_word_length = 0
current_length = 0

for ch in sentence:
    if ch == ' ':
        if current_length > 0:
            last_word_length = current_length
            current_length = 0
    else:
        current_length += 1

if current_length > 0:
    last_word_length = current_length

print("Length of last word:",last_word_length)
```

<strong style="font-size:25px;color:green">Reverse traversal</strong>

```python
sentence = input("Enter a sentence: ")
print("You Entered :",sentence)
i = 0

# Find number of characters manually
for ch in sentence:
    i += 1

i -= 1

# Ignore trailing spaces
while i >= 0 and sentence[i] == ' ':
    i -= 1

count = 0

# Count last word
while i >= 0 and sentence[i] != ' ':
    count += 1
    i -= 1

print("Length of last word:", count)
```

<strong style="font-size:25px;color:green">Recursive solution</strong>

```python
def last_word_length(sentence, i, count):
    if i < 0:
        return count

    if sentence[i] == ' ':
        if count == 0:
            return last_word_length(sentence, i - 1, count)
        return count

    return last_word_length(sentence, i - 1, count + 1)


sentence = input("Enter a sentence: ")
print("You Entered :",sentence)
size = 0

for ch in sentence:
    size += 1

result = last_word_length(sentence, size - 1, 0)

print("Length of last word:", result)
```
