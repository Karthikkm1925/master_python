<strong style="font-size:30px;color:green">Capitalize the first letter of each word in a sentence</strong>

<strong style="font-size:25px;color:green">Using a boolean flag</strong>

```python
sentence = "hello world python programming"
print("You Entered :", sentence)

result = ""
is_first = True

for ch in sentence:
    if ch != " ":
        if is_first == True:
            if 'a' <= ch <= 'z':
                ch = chr(ord(ch) - 32)
            is_first = False
        result = result + ch
    else:
        is_first = True
        result = result + ch

print("Capitalized Sentence :", result)
```

<strong style="font-size:25px;color:green">Using index and previous character check</strong>

```python
sentence = "hello world python programming"
print("You Entered :", sentence)

result = ""

for i in range(len(sentence)):
    ch = sentence[i]

    if ch != " ":
        if i == 0 or sentence[i - 1] == " ":
            if 'a' <= ch <= 'z':
                ch = chr(ord(ch) - 32)

    result = result + ch

print("Capitalized Sentence :", result)
```

<strong style="font-size:25px;color:green">Manual word extraction and reconstruction</strong>

```python
sentence = "hello world python programming"
print("You Entered :", sentence)

words = []
word = ""

for ch in sentence:
    if ch != " ":
        word = word + ch
    else:
        if word != "":
            words.append(word)
            word = ""

if word != "":
    words.append(word)

capitalized_words = []

for word in words:
    first_char = word[0]
    if 'a' <= first_char <= 'z':
        first_char = chr(ord(first_char) - 32)

    rest_of_word = ""
    for i in range(1, len(word)):
        rest_of_word = rest_of_word + word[i]

    capitalized_words.append(first_char + rest_of_word)

result = ""
for i in range(len(capitalized_words)):
    if i == 0:
        result = capitalized_words[i]
    else:
        result = result + " " + capitalized_words[i]

print("Capitalized Sentence :", result)
```

<strong style="font-size:25px;color:green">Using a while loop</strong>

```python
sentence = "hello world python programming"
print("You Entered :", sentence)

result = ""
i = 0

while i < len(sentence):
    ch = sentence[i]

    if ch != " ":
        if i == 0 or sentence[i - 1] == " ":
            if 'a' <= ch <= 'z':
                ch = chr(ord(ch) - 32)

    result = result + ch
    i = i + 1

print("Capitalized Sentence :", result)
```

<strong style="font-size:25px;color:green">Using a custom function for capitalization</strong>

```python
def capitalize_char(ch):
    if 'a' <= ch <= 'z':
        return chr(ord(ch) - 32)
    return ch

sentence = "hello world python programming"
print("You Entered :", sentence)

result = ""
new_word = True

for ch in sentence:
    if ch == " ":
        new_word = True
        result = result + ch
    else:
        if new_word == True:
            result = result + capitalize_char(ch)
            new_word = False
        else:
            result = result + ch

print("Capitalized Sentence :", result)
```
