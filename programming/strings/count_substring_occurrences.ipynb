<strong style="font-size:30px;color:green">Count occurrences of a substring in a string</strong>

<strong style="font-size:25px;color:green">Using string slicing in a for loop</strong>

```python
main_str = "ababa"
sub_str = "aba"
print("Main String :", main_str)
print("Substring   :", sub_str)

count = 0
main_len = len(main_str)
sub_len = len(sub_str)

for i in range(main_len - sub_len + 1):
    if main_str[i : i + sub_len] == sub_str:
        count = count + 1

print("Occurrences :", count)
```

<strong style="font-size:25px;color:green">Using nested loops (Character-by-character check)</strong>

```python
main_str = "hello world"
sub_str = "python"
print("Main String :", main_str)
print("Substring   :", sub_str)

count = 0
main_len = len(main_str)
sub_len = len(sub_str)

for i in range(main_len - sub_len + 1):
    match = True
    for j in range(sub_len):
        if main_str[i + j] != sub_str[j]:
            match = False
            break
    if match == True:
        count = count + 1

print("Occurrences :", count)
```

<strong style="font-size:25px;color:green">Using a while loop for non-overlapping occurrences</strong>

```python
main_str = "banana"
sub_str = "an"
print("Main String :", main_str)
print("Substring   :", sub_str)

count = 0
main_len = len(main_str)
sub_len = len(sub_str)
i = 0

while i <= main_len - sub_len:
    match = True
    for j in range(sub_len):
        if main_str[i + j] != sub_str[j]:
            match = False
            break

    if match == True:
        count = count + 1
        i = i + sub_len
    else:
        i = i + 1

print("Occurrences (Non-Overlapping) :", count)
```

<strong style="font-size:25px;color:green">Using a custom helper function</strong>

```python
def is_match(main_str, sub_str, start_index):
    for j in range(len(sub_str)):
        if main_str[start_index + j] != sub_str[j]:
            return False
    return True

main_str = "mississippi"
sub_str = "iss"
print("Main String :", main_str)
print("Substring   :", sub_str)

count = 0
main_len = len(main_str)
sub_len = len(sub_str)

for i in range(main_len - sub_len + 1):
    if is_match(main_str, sub_str, i):
        count = count + 1

print("Occurrences :", count)
```

<strong style="font-size:25px;color:green">Using recursion</strong>

```python
def count_substring_recursive(main_str, sub_str, index):
    main_len = len(main_str)
    sub_len = len(sub_str)

    if index > main_len - sub_len:
        return 0

    match = True
    for j in range(sub_len):
        if main_str[index + j] != sub_str[j]:
            match = False
            break

    if match == True:
        return 1 + count_substring_recursive(main_str, sub_str, index + 1)
    else:
        return count_substring_recursive(main_str, sub_str, index + 1)

main_str = "programming"
sub_str = "code"
print("Main String :", main_str)
print("Substring   :", sub_str)

result = count_substring_recursive(main_str, sub_str, 0)
print("Occurrences :", result)
```
