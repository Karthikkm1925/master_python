<strong style="font-size:30px;color:green">Find the shortest word in a sentence</strong>

<strong style="font-size:25px;color:green">Using split() and min() with key=len</strong>

```python
sentence = "Python is a great programming language"
print("You Entered :", sentence)

words = sentence.split()
shortest_word = min(words, key=len)

print("Shortest Word :", shortest_word)
```

<strong style="font-size:25px;color:green">Single pass manual word tracking</strong>

```python
sentence = "Python is a great programming language"
print("You Entered :", sentence)

shortest_word = ""
current_word = ""

for ch in sentence + " ":
    if ch != " ":
        current_word += ch
    else:
        if current_word != "":
            if shortest_word == "" or len(current_word) < len(shortest_word):
                shortest_word = current_word
            current_word = ""

print("Shortest Word :", shortest_word)
```

<strong style="font-size:25px;color:green">Extracting words into a list and iterating</strong>

```python
sentence = "Python is a great programming language"
print("You Entered :", sentence)

words = sentence.split()
shortest_word = words[0] if words else ""

for word in words[1:]:
    if len(word) < len(shortest_word):
        shortest_word = word

print("Shortest Word :", shortest_word)
```

<strong style="font-size:25px;color:green">Using sorting with key=len</strong>

```python
sentence = "Python is a great programming language"
print("You Entered :", sentence)

words = sentence.split()
sorted_words = sorted(words, key=len)
shortest_word = sorted_words[0] if sorted_words else ""

print("Shortest Word :", shortest_word)
```

<strong style="font-size:25px;color:green">Without using split() or built-in len()</strong>

```python
sentence = "Python is a great programming language"
print("You Entered :", sentence)

words = []
word = ""

for ch in sentence:
    if ch != " ":
        word += ch
    else:
        if word != "":
            words.append(word)
            word = ""

if word != "":
    words.append(word)

shortest_word = ""
min_length = float("inf")

for w in words:
    count = 0
    for _ in w:
        count += 1
    if count < min_length:
        min_length = count
        shortest_word = w

print("Shortest Word :", shortest_word)
```
