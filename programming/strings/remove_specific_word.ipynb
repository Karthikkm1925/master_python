<strong style="font-size:30px;color:green">Remove a specific word from a sentence</strong>

<strong style="font-size:25px;color:green">Using split() and list comprehension</strong>

```python
sentence = "I love Python programming and Python is great"
word_to_remove = "Python"
print("You Entered :", sentence)
print("Word to Remove :", word_to_remove)

words = sentence.split()
result = " ".join([word for word in words if word != word_to_remove])

print("Result :", result)
```

<strong style="font-size:25px;color:green">Using a for loop and list accumulation</strong>

```python
sentence = "I love Python programming and Python is great"
word_to_remove = "Python"
print("You Entered :", sentence)
print("Word to Remove :", word_to_remove)

words = sentence.split()
filtered_words = []

for word in words:
    if word != word_to_remove:
        filtered_words.append(word)

result = " ".join(filtered_words)

print("Result :", result)
```

<strong style="font-size:25px;color:green">Using replace() method</strong>

```python
sentence = "I love Python programming and Python is great"
word_to_remove = "Python"
print("You Entered :", sentence)
print("Word to Remove :", word_to_remove)

result = sentence.replace(word_to_remove, "").split()
result = " ".join(result)

print("Result :", result)
```

<strong style="font-size:25px;color:green">Using regular expressions (re.sub with word boundaries)</strong>

```python
import re

sentence = "I love Python programming and Python is great"
word_to_remove = "Python"
print("You Entered :", sentence)
print("Word to Remove :", word_to_remove)

pattern = r"\b" + re.escape(word_to_remove) + r"\b\s*"
result = re.sub(pattern, "", sentence).strip()

print("Result :", result)
```

<strong style="font-size:25px;color:green">Without using split() or replace()</strong>

```python
sentence = "I love Python programming and Python is great"
word_to_remove = "Python"
print("You Entered :", sentence)
print("Word to Remove :", word_to_remove)

result = ""
current_word = ""

for ch in sentence:
    if ch != " ":
        current_word += ch
    else:
        if current_word != "" and current_word != word_to_remove:
            if result != "":
                result += " "
            result += current_word
        current_word = ""

if current_word != "" and current_word != word_to_remove:
    if result != "":
        result += " "
    result += current_word

print("Result :", result)
```
