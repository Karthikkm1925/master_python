<strong style="font-size:30px;color:green">Frequency Of charecters in given string</strong>

<strong style="font-size:25px;color:green">Using a frequency dictionary</strong>

```python
s = "programming"

freq = {}

for ch in s:
    if ch in freq:
        freq[ch] = freq[ch] + 1
    else:
        freq[ch] = 1

most = ""
highest = 0

for ch in freq:
    if freq[ch] > highest:
        highest = freq[ch]
        most = ch

print("Letter \'",most,"\' is most repeated")
```
