<strong style="font-size:30px;color:green">Count overlapping occurrences of a substring in a string</strong>

<strong style="font-size:25px;color:green">Using string slicing with a for loop</strong>

```python
main_str = "ababa"
sub_str = "aba"
print("Main String :", main_str)
print("Substring   :", sub_str)

count = 0
main_len = len(main_str)
sub_len = len(sub_str)

for i in range(main_len - sub_len + 1):
    if main_str[i : i + sub_len] == sub_str:
        count += 1

print("Overlapping Occurrences :", count)
```

<strong style="font-size:25px;color:green">Using str.startswith() and sum()</strong>

```python
main_str = "ababa"
sub_str = "aba"
print("Main String :", main_str)
print("Substring   :", sub_str)

main_len = len(main_str)
sub_len = len(sub_str)

count = sum(1 for i in range(main_len - sub_len + 1) if main_str.startswith(sub_str, i))

print("Overlapping Occurrences :", count)
```

<strong style="font-size:25px;color:green">Using str.find() in a while loop</strong>

```python
main_str = "ababa"
sub_str = "aba"
print("Main String :", main_str)
print("Substring   :", sub_str)

count = 0
start = 0

while True:
    start = main_str.find(sub_str, start)
    if start != -1:
        count += 1
        start += 1
    else:
        break

print("Overlapping Occurrences :", count)
```

<strong style="font-size:25px;color:green">Using regular expressions (re) with lookahead assertion</strong>

```python
import re

main_str = "ababa"
sub_str = "aba"
print("Main String :", main_str)
print("Substring   :", sub_str)

# Lookahead (?=...) matches without consuming characters to capture overlaps
matches = re.findall(f"(?={re.escape(sub_str)})", main_str)
count = len(matches)

print("Overlapping Occurrences :", count)
```

<strong style="font-size:25px;color:green">Without using built-in methods or libraries (Nested while loops)</strong>

```python
main_str = "ababa"
sub_str = "aba"
print("Main String :", main_str)
print("Substring   :", sub_str)

count = 0
main_len = len(main_str)
sub_len = len(sub_str)

i = 0
while i <= main_len - sub_len:
    match = True
    j = 0
    while j < sub_len:
        if main_str[i + j] != sub_str[j]:
            match = False
            break
        j += 1
    if match:
        count += 1
    i += 1

print("Overlapping Occurrences :", count)
```
