<strong style="font-size:30px;color:green">Check whether one string is a rotation of another string</strong>

<strong style="font-size:25px;color:green">Using string concatenation (s1 + s1)</strong>

```python
s1 = "waterbottle"
s2 = "erbottlewat"
print("First String :", s1)
print("Second String :", s2)

is_rotation = len(s1) == len(s2) and s2 in (s1 + s1)

if is_rotation:
    print("The strings are rotations of each other.")
else:
    print("The strings are not rotations of each other.")
```

<strong style="font-size:25px;color:green">Using loop with manual slicing</strong>

```python
s1 = "waterbottle"
s2 = "erbottlewat"
print("First String :", s1)
print("Second String :", s2)

is_rotation = False

if len(s1) == len(s2) and len(s1) > 0:
    for i in range(len(s1)):
        rotated = s1[i:] + s1[:i]
        if rotated == s2:
            is_rotation = True
            break

if is_rotation:
    print("The strings are rotations of each other.")
else:
    print("The strings are not rotations of each other.")
```

<strong style="font-size:25px;color:green">Using collections.deque rotation</strong>

```python
from collections import deque

s1 = "waterbottle"
s2 = "erbottlewat"
print("First String :", s1)
print("Second String :", s2)

is_rotation = False

if len(s1) == len(s2) and len(s1) > 0:
    d1 = deque(s1)
    d2 = deque(s2)
    for _ in range(len(s1)):
        d1.rotate(1)
        if d1 == d2:
            is_rotation = True
            break

if is_rotation:
    print("The strings are rotations of each other.")
else:
    print("The strings are not rotations of each other.")
```

<strong style="font-size:25px;color:green">Without in operator (manual substring search in s1 + s1)</strong>

```python
s1 = "waterbottle"
s2 = "erbottlewat"
print("First String :", s1)
print("Second String :", s2)

is_rotation = False

if len(s1) == len(s2) and len(s1) > 0:
    doubled = s1 + s1
    n = len(doubled)
    m = len(s2)

    for i in range(n - m + 1):
        match = True
        for j in range(m):
            if doubled[i + j] != s2[j]:
                match = False
                break
        if match:
            is_rotation = True
            break

if is_rotation:
    print("The strings are rotations of each other.")
else:
    print("The strings are not rotations of each other.")
```

<strong style="font-size:25px;color:green">Using list queue with pop(0) and append()</strong>

```python
s1 = "waterbottle"
s2 = "erbottlewat"
print("First String :", s1)
print("Second String :", s2)

is_rotation = False

if len(s1) == len(s2) and len(s1) > 0:
    q1 = list(s1)
    target = list(s2)

    for _ in range(len(s1)):
        q1.append(q1.pop(0))
        if q1 == target:
            is_rotation = True
            break

if is_rotation:
    print("The strings are rotations of each other.")
else:
    print("The strings are not rotations of each other.")
```
