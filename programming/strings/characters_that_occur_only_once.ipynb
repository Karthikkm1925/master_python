<strong style="font-size:30px;color:green">Find all characters that occur only once in a string</strong>

<strong style="font-size:25px;color:green">Using collections.Counter</strong>

```python
from collections import Counter

s = "programming"
print("You Entered :", s)

counts = Counter(s)
unique_once = [char for char, count in counts.items() if count == 1]

print("Characters occurring only once :", unique_once)
```

<strong style="font-size:25px;color:green">Using a frequency dictionary</strong>

```python
s = "programming"
print("You Entered :", s)

freq = {}
for char in s:
    freq[char] = freq.get(char, 0) + 1

unique_once = [char for char, count in freq.items() if count == 1]

print("Characters occurring only once :", unique_once)
```

<strong style="font-size:25px;color:green">Using str.count()</strong>

```python
s = "programming"
print("You Entered :", s)

unique_once = [char for char in s if s.count(char) == 1]

print("Characters occurring only once :", unique_once)
```

<strong style="font-size:25px;color:green">Using find() and rfind()</strong>

```python
s = "programming"
print("You Entered :", s)

unique_once = []
for char in s:
    if s.find(char) == s.rfind(char):
        unique_once.append(char)

print("Characters occurring only once :", unique_once)
```

<strong style="font-size:25px;color:green">Without using built-in methods, sets, or dictionaries</strong>

```python
s = "programming"
print("You Entered :", s)

# Track unique characters and counts using parallel lists
unique_chars = []
counts = []

for char in s:
    found = False
    for i in range(len(unique_chars)):
        if unique_chars[i] == char:
            counts[i] += 1
            found = True
            break
    if not found:
        unique_chars.append(char)
        counts.append(1)

# Filter characters with a count of exactly 1
unique_once = []
for i in range(len(counts)):
    if counts[i] == 1:
        unique_once.append(unique_chars[i])

print("Characters occurring only once :", unique_once)
```
