<strong style="font-size:30px;color:green">Check if a string is a valid IPv4 address</strong>

<strong style="font-size:25px;color:green">Using split() and basic loops</strong>

```python
ip = "192.168.1.1"
print("You Entered :", ip)

parts = ip.split(".")
is_valid = True

if len(parts) != 4:
    is_valid = False
else:
    for part in parts:
        if not part.isdigit():
            is_valid = False
            break
        if len(part) > 1 and part[0] == "0":
            is_valid = False
            break
        val = int(part)
        if val < 0 or val > 255:
            is_valid = False
            break

if is_valid:
    print("Valid IPv4 address")
else:
    print("Invalid IPv4 address")
```

<strong style="font-size:25px;color:green">Using ipaddress module</strong>

```python
import ipaddress

ip = "192.168.1.1"
print("You Entered :", ip)

try:
    ipaddress.IPv4Address(ip)
    print("Valid IPv4 address")
except ValueError:
    print("Invalid IPv4 address")
```

<strong style="font-size:25px;color:green">Using regular expression</strong>

```python
import re

ip = "192.168.1.1"
print("You Entered :", ip)

pattern = r"^((25[0-5]|2[0-4][0-9]|[01]?[0-9][0-9]?)\.){3}(25[0-5]|2[0-4][0-9]|[01]?[0-9][0-9]?)$"

if re.match(pattern, ip):
    print("Valid IPv4 address")
else:
    print("Invalid IPv4 address")
```

<strong style="font-size:25px;color:green">Without split() or built-in functions</strong>

```python
ip = "192.168.1.1"
print("You Entered :", ip)

is_valid = True
dot_count = 0
current_num = ""

for i in range(len(ip)):
    ch = ip[i]
    if ch == ".":
        dot_count += 1
        if len(current_num) == 0 or len(current_num) > 3:
            is_valid = False
            break
        if len(current_num) > 1 and current_num[0] == "0":
            is_valid = False
            break
        val = 0
        for digit in current_num:
            val = val * 10 + (ord(digit) - ord('0'))
        if val < 0 or val > 255:
            is_valid = False
            break
        current_num = ""
    else:
        if ch < "0" or ch > "9":
            is_valid = False
            break
        current_num += ch

if dot_count != 3:
    is_valid = False

if is_valid and current_num:
    if len(current_num) > 1 and current_num[0] == "0":
        is_valid = False
    else:
        val = 0
        for digit in current_num:
            val = val * 10 + (ord(digit) - ord('0'))
        if val < 0 or val > 255:
            is_valid = False

if is_valid:
    print("Valid IPv4 address")
else:
    print("Invalid IPv4 address")
```

<strong style="font-size:25px;color:green">Using socket module</strong>

```python
import socket

ip = "192.168.1.1"
print("You Entered :", ip)

try:
    socket.inet_aton(ip)
    if ip.count(".") == 3:
        print("Valid IPv4 address")
    else:
        print("Invalid IPv4 address")
except socket.error:
    print("Invalid IPv4 address")
```
