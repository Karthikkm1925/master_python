<strong style="font-size:30px;color:green">Check if a string is a valid URL</strong>

<strong style="font-size:25px;color:green">Using Basic Loops</strong>

```python
url = "https://www.google.com"
print("You Entered :", url)

valid = True
http = False
dot_found = False

if len(url) >= 8:

    if url[0] == "h" and url[1] == "t" and url[2] == "t" and url[3] == "p" and url[4] == "s" and url[5] == ":" and url[6] == "/" and url[7] == "/":
        http = True

    elif url[0] == "h" and url[1] == "t" and url[2] == "t" and url[3] == "p" and url[4] == ":" and url[5] == "/" and url[6] == "/":
        http = True

if not http:
    valid = False

if valid:

    for i in range(len(url)):
        if url[i] == ".":
            dot_found = True
            break

    if not dot_found:
        valid = False

if valid:
    print("Valid URL")
else:
    print("Invalid URL")
```

<strong style="font-size:25px;color:green">Without in, startswith() and endswith()</strong>

```python
url = "https://www.google.com"
print("You Entered :", url)

valid = True

if len(url) < 8:
    valid = False

if valid:

    if url[0] != "h":
        valid = False
    elif url[1] != "t":
        valid = False
    elif url[2] != "t":
        valid = False
    elif url[3] != "p":
        valid = False
    elif url[4] != "s":
        valid = False
    elif url[5] != ":":
        valid = False
    elif url[6] != "/":
        valid = False
    elif url[7] != "/":
        valid = False

if valid:

    dot_found = False

    for ch in url:
        if ch == ".":
            dot_found = True
            break

    if not dot_found:
        valid = False

if valid:
    print("Valid URL")
else:
    print("Invalid URL")
```

<strong style="font-size:25px;color:green">Checking Protocol and Domain Separately</strong>

```python
url = "https://www.google.com"
print("You Entered :", url)

valid = True
protocol_end = -1

if len(url) >= 8:

    if url[0] == "h" and url[1] == "t" and url[2] == "t" and url[3] == "p" and url[4] == "s":
        if url[5] == ":" and url[6] == "/" and url[7] == "/":
            protocol_end = 7

if protocol_end == -1:
    valid = False

if valid:

    domain = ""

    i = protocol_end + 1

    while i < len(url):
        domain += url[i]
        i += 1

    if len(domain) == 0:
        valid = False

if valid:

    dot_found = False

    for ch in domain:
        if ch == ".":
            dot_found = True
            break

    if not dot_found:
        valid = False

if valid:
    print("Valid URL")
else:
    print("Invalid URL")
```

<strong style="font-size:25px;color:green">Checking Every Character</strong>

```python
url = "https://www.google.com"
print("You Entered :", url)

valid = True

if len(url) < 8:
    valid = False

if valid:

    if url[0:8] != "https://":
        if url[0:7] != "http://":
            valid = False

if valid:

    start = 8

    if url[4] == ":":
        start = 7

    if start >= len(url):
        valid = False

if valid:

    dot_found = False

    i = start

    while i < len(url):

        if url[i] == ".":
            dot_found = True

        if url[i] == " ":
            valid = False

        i += 1

    if not dot_found:
        valid = False

if valid:
    print("Valid URL")
else:
    print("Invalid URL")
```

<strong style="font-size:25px;color:green">Strict Basic Validation</strong>

```python
url = "https://www.google.com"
print("You Entered :", url)

valid = True

if len(url) < 8:
    valid = False

if valid:

    if url[0] == "h" and url[1] == "t" and url[2] == "t" and url[3] == "p":

        if url[4] == "s":
            if url[5] == ":" and url[6] == "/" and url[7] == "/":
                start = 8
            else:
                valid = False

        elif url[4] == ":" and url[5] == "/" and url[6] == "/":
            start = 7

        else:
            valid = False

    else:
        valid = False

if valid:

    if start >= len(url):
        valid = False

if valid:

    dot_found = False
    previous_dot = False

    i = start

    while i < len(url):

        ch = url[i]

        if ch == " ":
            valid = False

        if ch == ".":
            dot_found = True

            if previous_dot:
                valid = False

            previous_dot = True
        else:
            previous_dot = False

        i += 1

    if not dot_found:
        valid = False

if valid:

    if url[len(url) - 1] == ".":
        valid = False

if valid:
    print("Valid URL")
else:
    print("Invalid URL")
```

<strong style="font-size:25px;color:green">Using Regular Expression</strong>

```python
import re

url = "https://www.google.com"
print("You Entered :", url)

pattern = r"^(https?://)[A-Za-z0-9.-]+\.[A-Za-z]{2,}(/.*)?$"

if re.match(pattern, url):
    print("Valid URL")
else:
    print("Invalid URL")
```
