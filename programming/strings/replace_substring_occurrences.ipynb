<strong style="font-size:30px;color:green">Replace all occurrences of a substring with another substring</strong>

<strong style="font-size:25px;color:green">Using string slicing in a while loop</strong>

```python
main_str = "hello world, hello python"
old_sub = "hello"
new_sub = "hi"
print("Main String   :", main_str)
print("Old Substring :", old_sub)
print("New Substring :", new_sub)

result_str = ""
main_len = len(main_str)
old_len = len(old_sub)
i = 0

if old_len > 0:
    while i < main_len:
        if i <= main_len - old_len and main_str[i : i + old_len] == old_sub:
            result_str = result_str + new_sub
            i = i + old_len
        else:
            result_str = result_str + main_str[i]
            i = i + 1
else:
    result_str = main_str

print("Replaced String:", result_str)
```

<strong style="font-size:25px;color:green">Using nested loops (Character-by-character check)</strong>

```python
main_str = "banana"
old_sub = "an"
new_sub = "on"
print("Main String   :", main_str)
print("Old Substring :", old_sub)
print("New Substring :", new_sub)

result_str = ""
main_len = len(main_str)
old_len = len(old_sub)
i = 0

if old_len > 0:
    while i < main_len:
        match = True
        if i <= main_len - old_len:
            for j in range(old_len):
                if main_str[i + j] != old_sub[j]:
                    match = False
                    break
        else:
            match = False

        if match == True:
            result_str = result_str + new_sub
            i = i + old_len
        else:
            result_str = result_str + main_str[i]
            i = i + 1
else:
    result_str = main_str

print("Replaced String:", result_str)
```

<strong style="font-size:25px;color:green">Using a custom helper function</strong>

```python
def check_match(main_str, sub_str, start_index):
    for j in range(len(sub_str)):
        if main_str[start_index + j] != sub_str[j]:
            return False
    return True

main_str = "cat and dog and cat"
old_sub = "cat"
new_sub = "lion"
print("Main String   :", main_str)
print("Old Substring :", old_sub)
print("New Substring :", new_sub)

result_str = ""
main_len = len(main_str)
old_len = len(old_sub)
i = 0

if old_len > 0:
    while i < main_len:
        if i <= main_len - old_len and check_match(main_str, old_sub, i):
            result_str = result_str + new_sub
            i = i + old_len
        else:
            result_str = result_str + main_str[i]
            i = i + 1
else:
    result_str = main_str

print("Replaced String:", result_str)
```

<strong style="font-size:25px;color:green">Using recursion</strong>

```python
def replace_substring_recursive(main_str, old_sub, new_sub, index=0):
    main_len = len(main_str)
    old_len = len(old_sub)

    if old_len == 0 or index >= main_len:
        return main_str[index:] if old_len > 0 else main_str

    if index <= main_len - old_len:
        match = True
        for j in range(old_len):
            if main_str[index + j] != old_sub[j]:
                match = False
                break
        if match == True:
            return new_sub + replace_substring_recursive(main_str, old_sub, new_sub, index + old_len)

    return main_str[index] + replace_substring_recursive(main_str, old_sub, new_sub, index + 1)

main_str = "abcabcabc"
old_sub = "abc"
new_sub = "xyz"
print("Main String   :", main_str)
print("Old Substring :", old_sub)
print("New Substring :", new_sub)

result = replace_substring_recursive(main_str, old_sub, new_sub)
print("Replaced String:", result)
```
