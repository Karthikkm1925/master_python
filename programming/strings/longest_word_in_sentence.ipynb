<strong style="font-size:30px;color:green">Find the longest word in a sentence</strong>

<strong style="font-size:25px;color:green">Single pass manual word tracking</strong>

```python
sentence = "The quick brown fox jumps over the lazy dog"
print("You Entered :", sentence)

longest_word = ""
current_word = ""

for ch in sentence + " ":
    if ch != " ":
        current_word = current_word + ch
    else:
        if len(current_word) > len(longest_word):
            longest_word = current_word
        current_word = ""

print("Longest Word :", longest_word)
```

<strong style="font-size:25px;color:green">Extracting words into a list and comparing</strong>

```python
sentence = "The quick brown fox jumps over the lazy dog"
print("You Entered :", sentence)

words = []
word = ""

for ch in sentence:
    if ch != " ":
        word = word + ch
    else:
        if word != "":
            words.append(word)
            word = ""

if word != "":
    words.append(word)

longest_word = ""
max_length = 0

for w in words:
    count = 0
    for ch in w:
        count = count + 1

    if count > max_length:
        max_length = count
        longest_word = w

print("Longest Word :", longest_word)
```

<strong style="font-size:25px;color:green">Using comparison with nested loops</strong>

```python
sentence = "The quick brown fox jumps over the lazy dog"
print("You Entered :", sentence)

words = []
word = ""

for ch in sentence:
    if ch != " ":
        word = word + ch
    else:
        if word != "":
            words.append(word)
            word = ""

if word != "":
    words.append(word)

longest_word = ""

for i in range(len(words)):
    is_longest = True
    for j in range(len(words)):
        if len(words[j]) > len(words[i]):
            is_longest = False
            break
    if is_longest == True:
        longest_word = words[i]

print("Longest Word :", longest_word)
```

<strong style="font-size:25px;color:green">Using a while loop</strong>

```python
sentence = "The quick brown fox jumps over the lazy dog"
print("You Entered :", sentence)

longest_word = ""
current_word = ""
i = 0

while i < len(sentence):
    ch = sentence[i]

    if ch != " ":
        current_word = current_word + ch
    else:
        if len(current_word) > len(longest_word):
            longest_word = current_word
        current_word = ""

    i = i + 1

if len(current_word) > len(longest_word):
    longest_word = current_word

print("Longest Word :", longest_word)
```

<strong style="font-size:25px;color:green">Using a custom function</strong>

```python
def find_longest(sentence):
    longest = ""
    current = ""

    for ch in sentence + " ":
        if ch != " ":
            current = current + ch
        else:
            if len(current) > len(longest):
                longest = current
            current = ""

    return longest

sentence = "The quick brown fox jumps over the lazy dog"
print("You Entered :", sentence)

result = find_longest(sentence)
print("Longest Word :", result)
```
