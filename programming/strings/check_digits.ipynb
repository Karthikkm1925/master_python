<strong style="font-size:30px;color:green">Check if a string contains only digits</strong>

<strong style="font-size:25px;color:green">Using character range comparison</strong>

```python
string = "12345"
print("You Entered :", string)
is_digit = True

if len(string) == 0:
    is_digit = False
else:
    for char in string:
        if char < '0' or char > '9':
            is_digit = False
            break

if is_digit == True:
    print("The string contains only digits.")
else:
    print("The string does not contain only digits.")
```

<strong style="font-size:25px;color:green">Using a reference string of digits</strong>

```python
string = "12345"
print("You Entered :", string)
digits = "0123456789"
is_digit = True

if len(string) == 0:
    is_digit = False
else:
    for char in string:
        if char not in digits:
            is_digit = False
            break

if is_digit == True:
    print("The string contains only digits.")
else:
    print("The string does not contain only digits.")
```

<strong style="font-size:25px;color:green">Using nested loops</strong>

```python
string = "12345"
print("You Entered :", string)
digits = "0123456789"
is_digit = True

if len(string) == 0:
    is_digit = False
else:
    for i in range(len(string)):
        found = False
        for j in range(len(digits)):
            if string[i] == digits[j]:
                found = True
                break
        if found == False:
            is_digit = False
            break

if is_digit == True:
    print("The string contains only digits.")
else:
    print("The string does not contain only digits.")
```

<strong style="font-size:25px;color:green">Using a while loop</strong>

```python
string = "12345"
print("You Entered :", string)
is_digit = True
i = 0

if len(string) == 0:
    is_digit = False
else:
    while i < len(string):
        if string[i] < '0' or string[i] > '9':
            is_digit = False
            break
        i = i + 1

if is_digit == True:
    print("The string contains only digits.")
else:
    print("The string does not contain only digits.")
```

<strong style="font-size:25px;color:green">Using ASCII values</strong>

```python
string = "12345"
print("You Entered :", string)
is_digit = True

if len(string) == 0:
    is_digit = False
else:
    for char in string:
        ascii_val = ord(char)
        if ascii_val < 48 or ascii_val > 57:
            is_digit = False
            break

if is_digit == True:
    print("The string contains only digits.")
else:
    print("The string does not contain only digits.")
```

<strong style="font-size:25px;color:green">Using digit count comparison</strong>

```python
string = "12345"
print("You Entered :", string)
digit_count = 0

for char in string:
    if '0' <= char <= '9':
        digit_count = digit_count + 1

if len(string) > 0 and digit_count == len(string):
    print("The string contains only digits.")
else:
    print("The string does not contain only digits.")
```

<strong style="font-size:25px;color:green">Using recursion</strong>

```python
def check_digits(string, index):
    if len(string) == 0:
        return False
    if index == len(string):
        return True
    if string[index] < '0' or string[index] > '9':
        return False
    return check_digits(string, index + 1)

string = "12345"
print("You Entered :", string)
result = check_digits(string, 0)

if result == True:
    print("The string contains only digits.")
else:
    print("The string does not contain only digits.")
```
