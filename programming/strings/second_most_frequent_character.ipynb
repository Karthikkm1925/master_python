<strong style="font-size:30px;color:green">Find the second most frequent character in a string</strong>

<strong style="font-size:25px;color:green">Using frequency dictionary and sorting</strong>

```python
s = "success"
print("You Entered :", s)

freq = {}
for ch in s:
    freq[ch] = freq.get(ch, 0) + 1

sorted_freq = sorted(freq.items(), key=lambda x: x[1], reverse=True)

if len(sorted_freq) >= 2:
    print("Second most frequent character:", sorted_freq[1][0])
else:
    print("Not enough unique characters.")
```

<strong style="font-size:25px;color:green">Using collections.Counter</strong>

```python
from collections import Counter

s = "success"
print("You Entered :", s)

counts = Counter(s).most_common()

if len(counts) >= 2:
    print("Second most frequent character:", counts[1][0])
else:
    print("Not enough unique characters.")
```

<strong style="font-size:25px;color:green">Without built-in sorting</strong>

```python
s = "success"
print("You Entered :", s)

freq = {}
for ch in s:
    freq[ch] = freq.get(ch, 0) + 1

first_char, second_char = None, None
first_count, second_count = 0, 0

for ch, count in freq.items():
    if count > first_count:
        second_count = first_count
        second_char = first_char
        first_count = count
        first_char = ch
    elif count > second_count and count < first_count:
        second_count = count
        second_char = ch

if second_char:
    print("Second most frequent character:", second_char)
else:
    print("Not enough unique characters.")
```

<strong style="font-size:25px;color:green">Without built-in functions or dictionary</strong>

```python
s = "success"
print("You Entered :", s)

unique_chars = []
counts = []

for ch in s:
    found = False
    for i in range(len(unique_chars)):
        if unique_chars[i] == ch:
            counts[i] += 1
            found = True
            break
    if not found:
        unique_chars.append(ch)
        counts.append(1)

first_count, second_count = 0, 0
second_char = None

for i in range(len(counts)):
    if counts[i] > first_count:
        second_count = first_count
        first_count = counts[i]
    elif counts[i] > second_count and counts[i] < first_count:
        second_count = counts[i]

for i in range(len(counts)):
    if counts[i] == second_count:
        second_char = unique_chars[i]
        break

if second_char:
    print("Second most frequent character:", second_char)
else:
    print("Not enough unique characters.")
```

<strong style="font-size:25px;color:green">Using heapq module</strong>

```python
import heapq
from collections import Counter

s = "success"
print("You Entered :", s)

freq = Counter(s)

top_two = heapq.nlargest(2, freq.items(), key=lambda x: x[1])

if len(top_two) >= 2:
    print("Second most frequent character:", top_two[1][0])
else:
    print("Not enough unique characters.")
```
