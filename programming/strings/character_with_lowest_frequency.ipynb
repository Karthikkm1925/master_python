<strong style="font-size:30px;color:green">Find the character with the lowest frequency</strong>

<strong style="font-size:25px;color:green">Using collections.Counter and most_common()</strong>

```python
from collections import Counter

s = "aabbbcccc"
print("You Entered :", s)

char_counts = Counter(s)
# most_common() returns elements sorted by frequency descending; the last element has the lowest frequency
char, count = char_counts.most_common()[-1]

print("Lowest Frequency Character :", char)
print("Count :", count)
```

<strong style="font-size:25px;color:green">Using a frequency dictionary and min()</strong>

```python
s = "aabbbcccc"
print("You Entered :", s)

freq = {}
for char in s:
    freq[char] = freq.get(char, 0) + 1

min_char = min(freq, key=freq.get)
min_count = freq[min_char]

print("Lowest Frequency Character :", min_char)
print("Count :", min_count)
```

<strong style="font-size:25px;color:green">Using set() and str.count()</strong>

```python
s = "aabbbcccc"
print("You Entered :", s)

unique_chars = set(s)
min_char = min(unique_chars, key=s.count)
min_count = s.count(min_char)

print("Lowest Frequency Character :", min_char)
print("Count :", min_count)
```

<strong style="font-size:25px;color:green">Tracking minimum frequency by iterating over dictionary items</strong>

```python
s = "aabbbcccc"
print("You Entered :", s)

freq = {}
for char in s:
    freq[char] = freq.get(char, 0) + 1

min_char = None
min_count = float("inf")

for char, count in freq.items():
    if count < min_count:
        min_count = count
        min_char = char

print("Lowest Frequency Character :", min_char)
print("Count :", min_count)
```

<strong style="font-size:25px;color:green">Without using built-in functions, methods, sets, or dictionaries</strong>

```python
s = "aabbbcccc"
print("You Entered :", s)

# Track unique characters and frequencies using parallel lists
unique_chars = []
counts = []

for char in s:
    found = False
    for i in range(len(unique_chars)):
        if unique_chars[i] == char:
            counts[i] += 1
            found = True
            break
    if not found:
        unique_chars.append(char)
        counts.append(1)

# Find the minimum frequency and corresponding character manually
min_count = counts[0]
min_char = unique_chars[0]

for i in range(1, len(counts)):
    if counts[i] < min_count:
        min_count = counts[i]
        min_char = unique_chars[i]

print("Lowest Frequency Character :", min_char)
print("Count :", min_count)
```
