<strong style="font-size:30px;color:green">Find the character with the highest frequency and its count</strong>

<strong style="font-size:25px;color:green">Using collections.Counter and most_common()</strong>

```python
from collections import Counter

s = "abracadabra"
print("You Entered :", s)

char_counts = Counter(s)
char, count = char_counts.most_common(1)[0]

print("Highest Frequency Character :", char)
print("Count :", count)
```

<strong style="font-size:25px;color:green">Using a frequency dictionary and max()</strong>

```python
s = "abracadabra"
print("You Entered :", s)

freq = {}
for char in s:
    freq[char] = freq.get(char, 0) + 1

max_char = max(freq, key=freq.get)
max_count = freq[max_char]

print("Highest Frequency Character :", max_char)
print("Count :", max_count)
```

<strong style="font-size:25px;color:green">Using set() and str.count()</strong>

```python
s = "abracadabra"
print("You Entered :", s)

unique_chars = set(s)
max_char = max(unique_chars, key=s.count)
max_count = s.count(max_char)

print("Highest Frequency Character :", max_char)
print("Count :", max_count)
```

<strong style="font-size:25px;color:green">Tracking maximum frequency in a single pass</strong>

```python
s = "abracadabra"
print("You Entered :", s)

freq = {}
max_char = ""
max_count = 0

for char in s:
    freq[char] = freq.get(char, 0) + 1
    if freq[char] > max_count:
        max_count = freq[char]
        max_char = char

print("Highest Frequency Character :", max_char)
print("Count :", max_count)
```

<strong style="font-size:25px;color:green">Without using built-in functions, methods, sets, or dictionaries</strong>

```python
s = "abracadabra"
print("You Entered :", s)

# Track unique characters and frequencies using parallel lists
unique_chars = []
counts = []

for char in s:
    found = False
    for i in range(len(unique_chars)):
        if unique_chars[i] == char:
            counts[i] += 1
            found = True
            break
    if not found:
        unique_chars.append(char)
        counts.append(1)

# Find the maximum frequency and corresponding character
max_count = 0
max_char = ""
for i in range(len(counts)):
    if counts[i] > max_count:
        max_count = counts[i]
        max_char = unique_chars[i]

print("Highest Frequency Character :", max_char)
print("Count :", max_count)
```
