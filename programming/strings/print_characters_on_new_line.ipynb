<strong style="font-size:30px;color:green">Print each character of a string on a new line</strong>

<strong style="font-size:25px;color:green">Using a for loop</strong>

```python
s = "Python"
print("You Entered :", s)

for ch in s:
    print(ch)
```

<strong style="font-size:25px;color:green">Using a while loop with indexing</strong>

```python
s = "Python"
print("You Entered :", s)

i = 0
while i < len(s):
    print(s[i])
    i += 1
```

<strong style="font-size:25px;color:green">Using join() with newline separator</strong>

```python
s = "Python"
print("You Entered :", s)

print("\n".join(s))
```

<strong style="font-size:25px;color:green">Using unpack operator (*s) with print(sep='\n')</strong>

```python
s = "Python"
print("You Entered :", s)

print(*s, sep="\n")
```

<strong style="font-size:25px;color:green">Using recursion</strong>

```python
def print_chars(string):
    if not string:
        return
    print(string[0])
    print_chars(string[1:])


s = "Python"
print("You Entered :", s)

print_chars(s)
```
