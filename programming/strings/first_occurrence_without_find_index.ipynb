<strong style="font-size:30px;color:green">Find the first occurrence of a specific character in a string without using find() or index()</strong>

<strong style="font-size:25px;color:green">Using a for loop with range()</strong>

```python
s = "programming"
target = "r"
print("You Entered :", s)
print("Target Character :", target)

found_index = -1
for i in range(len(s)):
    if s[i] == target:
        found_index = i
        break

if found_index != -1:
    print("First occurrence index:", found_index)
else:
    print("Character not found.")
```

<strong style="font-size:25px;color:green">Using enumerate()</strong>

```python
s = "programming"
target = "r"
print("You Entered :", s)
print("Target Character :", target)

found_index = -1
for index, ch in enumerate(s):
    if ch == target:
        found_index = index
        break

if found_index != -1:
    print("First occurrence index:", found_index)
else:
    print("Character not found.")
```

<strong style="font-size:25px;color:green">Using a while loop</strong>

```python
s = "programming"
target = "r"
print("You Entered :", s)
print("Target Character :", target)

found_index = -1
i = 0
while i < len(s):
    if s[i] == target:
        found_index = i
        break
    i += 1

if found_index != -1:
    print("First occurrence index:", found_index)
else:
    print("Character not found.")
```

<strong style="font-size:25px;color:green">Using next() and generator expression</strong>

```python
s = "programming"
target = "r"
print("You Entered :", s)
print("Target Character :", target)

found_index = next((i for i, ch in enumerate(s) if ch == target), -1)

if found_index != -1:
    print("First occurrence index:", found_index)
else:
    print("Character not found.")
```

<strong style="font-size:25px;color:green">Using recursion</strong>

```python
def find_first_occurrence(s, target, index=0):
    if index >= len(s):
        return -1
    if s[index] == target:
        return index
    return find_first_occurrence(s, target, index + 1)


s = "programming"
target = "r"
print("You Entered :", s)
print("Target Character :", target)

found_index = find_first_occurrence(s, target)

if found_index != -1:
    print("First occurrence index:", found_index)
else:
    print("Character not found.")
```
