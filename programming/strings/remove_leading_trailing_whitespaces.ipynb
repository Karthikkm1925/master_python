<strong style="font-size:30px;color:green">Remove all leading and trailing whitespaces from a string</strong>

<strong style="font-size:25px;color:green">Using strip() method</strong>

```python
s = "   Hello World!   "
print("You Entered :", repr(s))

result = s.strip()

print("Result :", repr(result))
```

<strong style="font-size:25px;color:green">Using lstrip() and rstrip() methods</strong>

```python
s = "   Hello World!   "
print("You Entered :", repr(s))

result = s.lstrip().rstrip()

print("Result :", repr(result))
```

<strong style="font-size:25px;color:green">Using two pointers and slicing</strong>

```python
s = "   Hello World!   "
print("You Entered :", repr(s))

start = 0
end = len(s) - 1

while start <= end and s[start] == " ":
    start += 1

while end >= start and s[end] == " ":
    end -= 1

result = s[start:end + 1]

print("Result :", repr(result))
```

<strong style="font-size:25px;color:green">Without built-in functions or slicing</strong>

```python
s = "   Hello World!   "
print("You Entered :", repr(s))

start = 0
end = len(s) - 1

while start <= end and s[start] == " ":
    start += 1

while end >= start and s[end] == " ":
    end -= 1

result = ""

for i in range(start, end + 1):
    result += s[i]

print("Result :", repr(result))
```

<strong style="font-size:25px;color:green">Using regular expressions</strong>

```python
import re

s = "   Hello World!   "
print("You Entered :", repr(s))

result = re.sub(r"^\s+|\s+$", "", s)

print("Result :", repr(result))
```
