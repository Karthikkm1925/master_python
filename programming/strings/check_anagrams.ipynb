<strong style="font-size:30px;color:green">Check if two strings are anagrams</strong>

<strong style="font-size:25px;color:green">Using two dictionaries</strong>

```python
string1 = "listen"
string2 = "silent"
print("String 1 :", string1)
print("String 2 :", string2)

is_anagram = True

if len(string1) != len(string2):
    is_anagram = False
else:
    count1 = {}
    count2 = {}

    for ch in string1:
        if ch in count1:
            count1[ch] = count1[ch] + 1
        else:
            count1[ch] = 1

    for ch in string2:
        if ch in count2:
            count2[ch] = count2[ch] + 1
        else:
            count2[ch] = 1

    if count1 != count2:
        is_anagram = False

if is_anagram:
    print("The strings are anagrams.")
else:
    print("The strings are not anagrams.")
```

<strong style="font-size:25px;color:green">Using a single dictionary (Increment & Decrement)</strong>

```python
string1 = "hello"
string2 = "world"
print("String 1 :", string1)
print("String 2 :", string2)

is_anagram = True

if len(string1) != len(string2):
    is_anagram = False
else:
    freq = {}

    for ch in string1:
        if ch in freq:
            freq[ch] = freq[ch] + 1
        else:
            freq[ch] = 1

    for ch in string2:
        if ch in freq:
            freq[ch] = freq[ch] - 1
        else:
            is_anagram = False
            break

    if is_anagram == True:
        for key in freq:
            if freq[key] != 0:
                is_anagram = False
                break

if is_anagram:
    print("The strings are anagrams.")
else:
    print("The strings are not anagrams.")
```

<strong style="font-size:25px;color:green">Using an ASCII value frequency array</strong>

```python
string1 = "earth"
string2 = "heart"
print("String 1 :", string1)
print("String 2 :", string2)

is_anagram = True

if len(string1) != len(string2):
    is_anagram = False
else:
    counts = [0] * 256

    for ch in string1:
        ascii_val = ord(ch)
        counts[ascii_val] = counts[ascii_val] + 1

    for ch in string2:
        ascii_val = ord(ch)
        counts[ascii_val] = counts[ascii_val] - 1

    for c in counts:
        if c != 0:
            is_anagram = False
            break

if is_anagram:
    print("The strings are anagrams.")
else:
    print("The strings are not anagrams.")
```

<strong style="font-size:25px;color:green">Using manual Bubble Sort on character lists</strong>

```python
string1 = "aab"
string2 = "abb"
print("String 1 :", string1)
print("String 2 :", string2)

is_anagram = True

if len(string1) != len(string2):
    is_anagram = False
else:
    list1 = []
    for ch in string1:
        list1.append(ch)

    list2 = []
    for ch in string2:
        list2.append(ch)

    for i in range(len(list1)):
        for j in range(0, len(list1) - i - 1):
            if list1[j] > list1[j + 1]:
                temp = list1[j]
                list1[j] = list1[j + 1]
                list1[j + 1] = temp

    for i in range(len(list2)):
        for j in range(0, len(list2) - i - 1):
            if list2[j] > list2[j + 1]:
                temp = list2[j]
                list2[j] = list2[j + 1]
                list2[j + 1] = temp

    for i in range(len(list1)):
        if list1[i] != list2[i]:
            is_anagram = False
            break

if is_anagram:
    print("The strings are anagrams.")
else:
    print("The strings are not anagrams.")
```

<strong style="font-size:25px;color:green">Using manual Selection Sort on character lists</strong>

```python
string1 = "evil"
string2 = "vile"
print("String 1 :", string1)
print("String 2 :", string2)

is_anagram = True

if len(string1) != len(string2):
    is_anagram = False
else:
    list1 = []
    for ch in string1:
        list1.append(ch)

    list2 = []
    for ch in string2:
        list2.append(ch)

    for i in range(len(list1)):
        min_idx = i
        for j in range(i + 1, len(list1)):
            if list1[j] < list1[min_idx]:
                min_idx = j
        temp = list1[i]
        list1[i] = list1[min_idx]
        list1[min_idx] = temp

    for i in range(len(list2)):
        min_idx = i
        for j in range(i + 1, len(list2)):
            if list2[j] < list2[min_idx]:
                min_idx = j
        temp = list2[i]
        list2[i] = list2[min_idx]
        list2[min_idx] = temp

    for i in range(len(list1)):
        if list1[i] != list2[i]:
            is_anagram = False
            break

if is_anagram:
    print("The strings are anagrams.")
else:
    print("The strings are not anagrams.")
```

<strong style="font-size:25px;color:green">Using character matching and element removal</strong>

```python
string1 = "cat"
string2 = "cats"
print("String 1 :", string1)
print("String 2 :", string2)

is_anagram = True

if len(string1) != len(string2):
    is_anagram = False
else:
    list2 = []
    for ch in string2:
        list2.append(ch)

    for ch in string1:
        found = False
        match_index = -1
        for i in range(len(list2)):
            if list2[i] == ch:
                match_index = i
                found = True
                break

        if found == False:
            is_anagram = False
            break
        else:
            new_list2 = []
            for k in range(len(list2)):
                if k != match_index:
                    new_list2.append(list2[k])
            list2 = new_list2

    if len(list2) != 0:
        is_anagram = False

if is_anagram:
    print("The strings are anagrams.")
else:
    print("The strings are not anagrams.")
```

<strong style="font-size:25px;color:green">Using pure nested loops count comparison</strong>

```python
string1 = "apple"
string2 = "papel"
print("String 1 :", string1)
print("String 2 :", string2)

is_anagram = True

if len(string1) != len(string2):
    is_anagram = False
else:
    for i in range(len(string1)):
        ch = string1[i]

        already_checked = False
        for k in range(i):
            if string1[k] == ch:
                already_checked = True
                break

        if already_checked:
            continue

        count1 = 0
        for j in range(len(string1)):
            if string1[j] == ch:
                count1 = count1 + 1

        count2 = 0
        for j in range(len(string2)):
            if string2[j] == ch:
                count2 = count2 + 1

        if count1 != count2:
            is_anagram = False
            break

if is_anagram:
    print("The strings are anagrams.")
else:
    print("The strings are not anagrams.")
```

<strong style="font-size:25px;color:green">Using recursion</strong>

```python
def check_anagram_recursive(s1, s2_list):
    if len(s1) == 0 and len(s2_list) == 0:
        return True
    if len(s1) != len(s2_list):
        return False

    ch = s1[0]
    found = False
    match_index = -1

    for i in range(len(s2_list)):
        if s2_list[i] == ch:
            match_index = i
            found = True
            break

    if found == False:
        return False

    new_s2_list = []
    for k in range(len(s2_list)):
        if k != match_index:
            new_s2_list.append(s2_list[k])

    rem_s1 = ""
    for i in range(1, len(s1)):
        rem_s1 = rem_s1 + s1[i]

    return check_anagram_recursive(rem_s1, new_s2_list)

string1 = "python"
string2 = "java"
print("String 1 :", string1)
print("String 2 :", string2)

list2 = []
for ch in string2:
    list2.append(ch)

if check_anagram_recursive(string1, list2):
    print("The strings are anagrams.")
else:
    print("The strings are not anagrams.")
```

<strong style="font-size:25px;color:green">Case-insensitive anagram check without .lower()</strong>

```python
string1 = "Listen"
string2 = "Silent"
print("String 1 :", string1)
print("String 2 :", string2)

is_anagram = True

if len(string1) != len(string2):
    is_anagram = False
else:
    freq = {}

    for ch in string1:
        if 'A' <= ch <= 'Z':
            ch = chr(ord(ch) + 32)
        if ch in freq:
            freq[ch] = freq[ch] + 1
        else:
            freq[ch] = 1

    for ch in string2:
        if 'A' <= ch <= 'Z':
            ch = chr(ord(ch) + 32)
        if ch in freq:
            freq[ch] = freq[ch] - 1
        else:
            is_anagram = False
            break

    if is_anagram == True:
        for key in freq:
            if freq[key] != 0:
                is_anagram = False
                break

if is_anagram:
    print("The strings are anagrams.")
else:
    print("The strings are not anagrams.")
```
