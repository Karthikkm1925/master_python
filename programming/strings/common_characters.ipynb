<strong style="font-size:30px;color:green">Find the common characters between two strings</strong>

<strong style="font-size:25px;color:green">Using set intersection</strong>

```python
s1 = "apple"
s2 = "sample"
print("First String :", s1)
print("Second String :", s2)

common = sorted(list(set(s1) & set(s2)))

print("Common characters:", common)
```

<strong style="font-size:25px;color:green">Using loop and list tracking</strong>

```python
s1 = "apple"
s2 = "sample"
print("First String :", s1)
print("Second String :", s2)

common = []

for ch in s1:
    if ch in s2 and ch not in common:
        common.append(ch)

print("Common characters:", common)
```

<strong style="font-size:25px;color:green">Without built-in functions or set</strong>

```python
s1 = "apple"
s2 = "sample"
print("First String :", s1)
print("Second String :", s2)

common = []

for i in range(len(s1)):
    ch = s1[i]
    is_in_s2 = False
    for j in range(len(s2)):
        if s2[j] == ch:
            is_in_s2 = True
            break

    if is_in_s2:
        already_added = False
        for k in range(len(common)):
            if common[k] == ch:
                already_added = True
                break
        if not already_added:
            common.append(ch)

print("Common characters:", common)
```

<strong style="font-size:25px;color:green">Using list comprehension</strong>

```python
s1 = "apple"
s2 = "sample"
print("First String :", s1)
print("Second String :", s2)

common = sorted(list(set([ch for ch in s1 if ch in s2])))

print("Common characters:", common)
```

<strong style="font-size:25px;color:green">Using collections.Counter</strong>

```python
from collections import Counter

s1 = "apple"
s2 = "sample"
print("First String :", s1)
print("Second String :", s2)

c1 = Counter(s1)
c2 = Counter(s2)

common = list((c1 & c2).elements())

print("Common characters:", common)
```
