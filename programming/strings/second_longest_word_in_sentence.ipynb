<strong style="font-size:30px;color:green">Find the second longest word in a sentence</strong>

<strong style="font-size:25px;color:green">Using split() and sorting by length</strong>

```python
sentence = "Python is an amazing programming language"
print("You Entered :", sentence)

words = list(set(sentence.split()))
sorted_words = sorted(words, key=len, reverse=True)

if len(sorted_words) >= 2:
    print("Second Longest Word :", sorted_words[1])
else:
    print("Not enough unique words.")
```

<strong style="font-size:25px;color:green">Single pass tracking of longest and second longest</strong>

```python
sentence = "Python is an amazing programming language"
print("You Entered :", sentence)

words = sentence.split()
longest = ""
second_longest = ""

for word in words:
    if len(word) > len(longest):
        second_longest = longest
        longest = word
    elif len(word) > len(second_longest) and len(word) < len(longest):
        second_longest = word

print("Second Longest Word :", second_longest)
```

<strong style="font-size:25px;color:green">Using heapq.nlargest</strong>

```python
import heapq

sentence = "Python is an amazing programming language"
print("You Entered :", sentence)

words = list(set(sentence.split()))
top_words = heapq.nlargest(2, words, key=len)

if len(top_words) >= 2:
    print("Second Longest Word :", top_words[1])
else:
    print("Not enough unique words.")
```
