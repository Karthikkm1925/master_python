<strong style="font-size:30px;color:green">Print all duplicate characters in a string</strong>

<strong style="font-size:25px;color:green">Using a frequency dictionary</strong>

```python
s = "programming"
print("You Entered :", s)

freq = {}
for ch in s:
    freq[ch] = freq.get(ch, 0) + 1

duplicates = [ch for ch, count in freq.items() if count > 1]

print("Duplicate characters:", duplicates)
```

<strong style="font-size:25px;color:green">Using collections.Counter</strong>

```python
from collections import Counter

s = "programming"
print("You Entered :", s)

counts = Counter(s)
duplicates = [ch for ch, count in counts.items() if count > 1]

print("Duplicate characters:", duplicates)
```

<strong style="font-size:25px;color:green">Using sets for tracking</strong>

```python
s = "programming"
print("You Entered :", s)

seen = set()
duplicates = set()

for ch in s:
    if ch in seen:
        duplicates.add(ch)
    else:
        seen.add(ch)

print("Duplicate characters:", sorted(list(duplicates)))
```

<strong style="font-size:25px;color:green">Using count() method</strong>

```python
s = "programming"
print("You Entered :", s)

duplicates = []
for ch in s:
    if s.count(ch) > 1 and ch not in duplicates:
        duplicates.append(ch)

print("Duplicate characters:", duplicates)
```

<strong style="font-size:25px;color:green">Without built-in functions or sets</strong>

```python
s = "programming"
print("You Entered :", s)

duplicates = []

for i in range(len(s)):
    is_dup = False
    for j in range(i + 1, len(s)):
        if s[i] == s[j]:
            is_dup = True
            break
    if is_dup:
        already_added = False
        for k in range(len(duplicates)):
            if duplicates[k] == s[i]:
                already_added = True
                break
        if not already_added:
            duplicates.append(s[i])

print("Duplicate characters:", duplicates)
```
