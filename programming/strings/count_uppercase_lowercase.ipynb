<strong style="font-size:30px;color:green">Count uppercase and lowercase letters in a string</strong>

<strong style="font-size:25px;color:green">Using isupper() and islower()</strong>

```python
s = "Hello World! 123"
print("You Entered :", s)

upper_count = 0
lower_count = 0

for ch in s:
    if ch.isupper():
        upper_count += 1
    elif ch.islower():
        lower_count += 1

print("Uppercase count:", upper_count)
print("Lowercase count:", lower_count)
```

<strong style="font-size:25px;color:green">Using ASCII character range comparison</strong>

```python
s = "Hello World! 123"
print("You Entered :", s)

upper_count = 0
lower_count = 0

for ch in s:
    if "A" <= ch <= "Z":
        upper_count += 1
    elif "a" <= ch <= "z":
        lower_count += 1

print("Uppercase count:", upper_count)
print("Lowercase count:", lower_count)
```

<strong style="font-size:25px;color:green">Using sum() and generator expressions</strong>

```python
s = "Hello World! 123"
print("You Entered :", s)

upper_count = sum(1 for ch in s if ch.isupper())
lower_count = sum(1 for ch in s if ch.islower())

print("Uppercase count:", upper_count)
print("Lowercase count:", lower_count)
```

<strong style="font-size:25px;color:green">Using a dictionary counter</strong>

```python
s = "Hello World! 123"
print("You Entered :", s)

counts = {"upper": 0, "lower": 0}

for ch in s:
    if ch.isupper():
        counts["upper"] += 1
    elif ch.islower():
        counts["lower"] += 1

print("Uppercase count:", counts["upper"])
print("Lowercase count:", counts["lower"])
```

<strong style="font-size:25px;color:green">Using filter() and len()</strong>

```python
s = "Hello World! 123"
print("You Entered :", s)

upper_count = len(list(filter(str.isupper, s)))
lower_count = len(list(filter(str.islower, s)))

print("Uppercase count:", upper_count)
print("Lowercase count:", lower_count)
```
