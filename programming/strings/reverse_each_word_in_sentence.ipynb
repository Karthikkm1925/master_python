<strong style="font-size:30px;color:green">Reverse each word in a sentence while keeping the word order unchanged</strong>

<strong style="font-size:25px;color:green">Using split() and string slicing [::-1]</strong>

```python
sentence = "I love Python programming"
print("You Entered :", sentence)

words = sentence.split()
reversed_words = [word[::-1] for word in words]
result = " ".join(reversed_words)

print("Result :", result)
```

<strong style="font-size:25px;color:green">Using loop and manual word reversing</strong>

```python
sentence = "I love Python programming"
print("You Entered :", sentence)

words = sentence.split()
reversed_words = []

for word in words:
    rev = ""
    for ch in word:
        rev = ch + rev
    reversed_words.append(rev)

result = " ".join(reversed_words)

print("Result :", result)
```

<strong style="font-size:25px;color:green">Single pass without split() using character accumulation</strong>

```python
sentence = "I love Python programming"
print("You Entered :", sentence)

result = ""
current_word = ""

for ch in sentence:
    if ch != " ":
        current_word = ch + current_word
    else:
        result += current_word + " "
        current_word = ""

result += current_word

print("Result :", result)
```

<strong style="font-size:25px;color:green">Using list as a stack for each word</strong>

```python
sentence = "I love Python programming"
print("You Entered :", sentence)

result = ""
stack = []

for ch in sentence:
    if ch != " ":
        stack.append(ch)
    else:
        while stack:
            result += stack.pop()
        result += " "

while stack:
    result += stack.pop()

print("Result :", result)
```

<strong style="font-size:25px;color:green">Using map() and reversed() with join()</strong>

```python
sentence = "I love Python programming"
print("You Entered :", sentence)

result = " ".join(map(lambda w: "".join(reversed(w)), sentence.split()))

print("Result :", result)
```
