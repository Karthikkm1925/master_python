<strong style="font-size:30px;color:green">Count the number of words in a sentence without using split()</strong>

<strong style="font-size:25px;color:green">Using a state flag (in_word tracking)</strong>

```python
s = "Python is a great programming language"
print("You Entered :", s)

count = 0
in_word = False

for ch in s:
    if ch != " ":
        if not in_word:
            count += 1
            in_word = True
    else:
        in_word = False

print("Number of words:", count)
```

<strong style="font-size:25px;color:green">Using loop and building words</strong>

```python
s = "Python is a great programming language"
print("You Entered :", s)

count = 0
current_word = ""

for ch in s:
    if ch != " ":
        current_word += ch
    else:
        if current_word != "":
            count += 1
            current_word = ""

if current_word != "":
    count += 1

print("Number of words:", count)
```

<strong style="font-size:25px;color:green">Using while loop with index navigation</strong>

```python
s = "Python is a great programming language"
print("You Entered :", s)

count = 0
i = 0
n = len(s)

while i < n:
    while i < n and s[i] == " ":
        i += 1
    if i < n:
        count += 1
        while i < n and s[i] != " ":
            i += 1

print("Number of words:", count)
```

<strong style="font-size:25px;color:green">Using regular expressions (re.findall)</strong>

```python
import re

s = "Python is a great programming language"
print("You Entered :", s)

count = len(re.findall(r"\S+", s))

print("Number of words:", count)
```

<strong style="font-size:25px;color:green">Using recursion</strong>

```python
def count_words(string, in_word=False):
    if not string:
        return 0
    if string[0] != " ":
        if not in_word:
            return 1 + count_words(string[1:], True)
        return count_words(string[1:], True)
    else:
        return count_words(string[1:], False)


s = "Python is a great programming language"
print("You Entered :", s)

count = count_words(s)

print("Number of words:", count)
```
