<strong style="font-size:30px;color:green">Count the number of characters in a string without using len()</strong>

<strong style="font-size:25px;color:green">Using a for loop</strong>

```python
s = "Hello World"
print("You Entered :", s)

count = 0
for ch in s:
    count += 1

print("Number of characters:", count)
```

<strong style="font-size:25px;color:green">Using sum() and generator expression</strong>

```python
s = "Hello World"
print("You Entered :", s)

count = sum(1 for ch in s)

print("Number of characters:", count)
```

<strong style="font-size:25px;color:green">Using a while loop and string indexing</strong>

```python
s = "Hello World"
print("You Entered :", s)

count = 0
while True:
    try:
        _ = s[count]
        count += 1
    except IndexError:
        break

print("Number of characters:", count)
```

<strong style="font-size:25px;color:green">Using recursion</strong>

```python
def count_chars(string):
    if string == "":
        return 0
    return 1 + count_chars(string[1:])


s = "Hello World"
print("You Entered :", s)

count = count_chars(s)

print("Number of characters:", count)
```

<strong style="font-size:25px;color:green">Using enumerate()</strong>

```python
s = "Hello World"
print("You Entered :", s)

count = 0
for i, _ in enumerate(s, 1):
    count = i

print("Number of characters:", count)
```
