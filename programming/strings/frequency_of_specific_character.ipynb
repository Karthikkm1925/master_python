<strong style="font-size:30px;color:green">Find the frequency of a specific character in a string</strong>

<strong style="font-size:25px;color:green">Using count() method</strong>

```python
s = "programming"
target = "r"
print("You Entered :", s)
print("Target Character :", target)

count = s.count(target)

print(f"Frequency of '{target}':", count)
```

<strong style="font-size:25px;color:green">Using a for loop</strong>

```python
s = "programming"
target = "r"
print("You Entered :", s)
print("Target Character :", target)

count = 0
for ch in s:
    if ch == target:
        count += 1

print(f"Frequency of '{target}':", count)
```

<strong style="font-size:25px;color:green">Using collections.Counter</strong>

```python
from collections import Counter

s = "programming"
target = "r"
print("You Entered :", s)
print("Target Character :", target)

counts = Counter(s)
count = counts[target]

print(f"Frequency of '{target}':", count)
```

<strong style="font-size:25px;color:green">Using sum() and generator expression</strong>

```python
s = "programming"
target = "r"
print("You Entered :", s)
print("Target Character :", target)

count = sum(1 for ch in s if ch == target)

print(f"Frequency of '{target}':", count)
```

<strong style="font-size:25px;color:green">Using recursion</strong>

```python
def char_frequency(string, target):
    if not string:
        return 0
    match = 1 if string[0] == target else 0
    return match + char_frequency(string[1:], target)


s = "programming"
target = "r"
print("You Entered :", s)
print("Target Character :", target)

count = char_frequency(s, target)

print(f"Frequency of '{target}':", count)
```
