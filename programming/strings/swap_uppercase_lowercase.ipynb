<strong style="font-size:30px;color:green">Swap uppercase characters to lowercase and lowercase characters to uppercase</strong>

<strong style="font-size:25px;color:green">Using swapcase() method</strong>

```python
s = "Hello World! Python 3.11"
print("You Entered :", s)

result = s.swapcase()

print("Result :", result)
```

<strong style="font-size:25px;color:green">Using isupper() and islower() with upper() and lower()</strong>

```python
s = "Hello World! Python 3.11"
print("You Entered :", s)

result = ""
for ch in s:
    if ch.isupper():
        result += ch.lower()
    elif ch.islower():
        result += ch.upper()
    else:
        result += ch

print("Result :", result)
```

<strong style="font-size:25px;color:green">Using ASCII value manipulation (ord and chr)</strong>

```python
s = "Hello World! Python 3.11"
print("You Entered :", s)

result = ""
for ch in s:
    if "a" <= ch <= "z":
        result += chr(ord(ch) - 32)
    elif "A" <= ch <= "Z":
        result += chr(ord(ch) + 32)
    else:
        result += ch

print("Result :", result)
```

<strong style="font-size:25px;color:green">Using list comprehension and join()</strong>

```python
s = "Hello World! Python 3.11"
print("You Entered :", s)

result = "".join([ch.lower() if ch.isupper() else ch.upper() if ch.islower() else ch for ch in s])

print("Result :", result)
```

<strong style="font-size:25px;color:green">Using bitwise XOR operation</strong>

```python
s = "Hello World! Python 3.11"
print("You Entered :", s)

result = ""
for ch in s:
    if ("a" <= ch <= "z") or ("A" <= ch <= "Z"):
        result += chr(ord(ch) ^ 32)
    else:
        result += ch

print("Result :", result)
```
