<strong style="font-size:30px;color:green">Find all duplicate words in a sentence</strong>

<strong style="font-size:25px;color:green">Using collections.Counter</strong>

```python
from collections import Counter

sentence = "Python is great and Python is fun and easy to learn"
print("You Entered :", sentence)

words = sentence.split()
counts = Counter(words)
duplicates = [word for word, count in counts.items() if count > 1]

print("Duplicate words:", duplicates)
```

<strong style="font-size:25px;color:green">Using a frequency dictionary</strong>

```python
sentence = "Python is great and Python is fun and easy to learn"
print("You Entered :", sentence)

words = sentence.split()
freq = {}

for word in words:
    freq[word] = freq.get(word, 0) + 1

duplicates = [word for word, count in freq.items() if count > 1]

print("Duplicate words:", duplicates)
```

<strong style="font-size:25px;color:green">Using sets for tracking seen and duplicate words</strong>

```python
sentence = "Python is great and Python is fun and easy to learn"
print("You Entered :", sentence)

words = sentence.split()
seen = set()
duplicates = set()

for word in words:
    if word in seen:
        duplicates.add(word)
    else:
        seen.add(word)

print("Duplicate words:", sorted(list(duplicates)))
```

<strong style="font-size:25px;color:green">Using list count() method</strong>

```python
sentence = "Python is great and Python is fun and easy to learn"
print("You Entered :", sentence)

words = sentence.split()
duplicates = []

for word in words:
    if words.count(word) > 1 and word not in duplicates:
        duplicates.append(word)

print("Duplicate words:", duplicates)
```

<strong style="font-size:25px;color:green">Without using split() or built-in count()</strong>

```python
sentence = "Python is great and Python is fun and easy to learn"
print("You Entered :", sentence)

words = []
current_word = ""

for ch in sentence:
    if ch != " ":
        current_word += ch
    else:
        if current_word != "":
            words.append(current_word)
            current_word = ""

if current_word != "":
    words.append(current_word)

duplicates = []

for i in range(len(words)):
    for j in range(i + 1, len(words)):
        if words[i] == words[j] and words[i] not in duplicates:
            duplicates.append(words[i])
            break

print("Duplicate words:", duplicates)
```
