<strong style="font-size:30px;color:green">Check whether one string is a substring of another without using in</strong>

<strong style="font-size:25px;color:green">Using str.find() method</strong>

```python
main_str = "hello world"
sub_str = "world"
print("Main String :", main_str)
print("Substring   :", sub_str)

is_substring = main_str.find(sub_str) != -1

if is_substring:
    print("The substring is present in the main string.")
else:
    print("The substring is not present in the main string.")
```

<strong style="font-size:25px;color:green">Using string slicing with a for loop</strong>

```python
main_str = "hello world"
sub_str = "world"
print("Main String :", main_str)
print("Substring   :", sub_str)

is_substring = False
main_len = len(main_str)
sub_len = len(sub_str)

for i in range(main_len - sub_len + 1):
    if main_str[i : i + sub_len] == sub_str:
        is_substring = True
        break

if is_substring:
    print("The substring is present in the main string.")
else:
    print("The substring is not present in the main string.")
```

<strong style="font-size:25px;color:green">Using str.count() method</strong>

```python
main_str = "hello world"
sub_str = "world"
print("Main String :", main_str)
print("Substring   :", sub_str)

is_substring = main_str.count(sub_str) > 0

if is_substring:
    print("The substring is present in the main string.")
else:
    print("The substring is not present in the main string.")
```

<strong style="font-size:25px;color:green">Using str.index() with try-except block</strong>

```python
main_str = "hello world"
sub_str = "world"
print("Main String :", main_str)
print("Substring   :", sub_str)

try:
    main_str.index(sub_str)
    is_substring = True
except ValueError:
    is_substring = False

if is_substring:
    print("The substring is present in the main string.")
else:
    print("The substring is not present in the main string.")
```

<strong style="font-size:25px;color:green">Without using built-in methods or in keyword (Nested while loops)</strong>

```python
main_str = "hello world"
sub_str = "world"
print("Main String :", main_str)
print("Substring   :", sub_str)

main_len = len(main_str)
sub_len = len(sub_str)
is_substring = False

i = 0
while i <= main_len - sub_len:
    match = True
    j = 0
    while j < sub_len:
        if main_str[i + j] != sub_str[j]:
            match = False
            break
        j += 1
    if match:
        is_substring = True
        break
    i += 1

if is_substring:
    print("The substring is present in the main string.")
else:
    print("The substring is not present in the main string.")
```
