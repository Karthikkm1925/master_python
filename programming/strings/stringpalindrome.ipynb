<strong style="font-size:30px;color:green">String Palindrome</strong>

<strong style="font-size:25px;color:green">Using Two Pointers</strong>

```python
s = input("Enter string: ")
print("You Entered :",s)
length = 0
for ch in s:
    length += 1

left = 0
right = length - 1
flag = True

while left < right:
    if s[left] != s[right]:
        flag = False
        break

    left += 1
    right -= 1

if flag:
    print(s,"is Palindrome")
else:
    print(s,"is Not Palindrome")
```

<strong style="font-size:25px;color:green">Using Reverse String</strong>

```python
s = input("Enter string: ")
print("You Entered :",s)
reverse = ""

for ch in s:
    reverse = ch + reverse

if s == reverse:
    print(s,"is Palindrome")
else:
    print(s,"is Not Palindrome")
```

<strong style="font-size:25px;color:green">Using Recursion</strong>

```python
def palindrome(s, left, right):

    if left >= right:
        return True

    if s[left] != s[right]:
        return False

    return palindrome(s, left + 1, right - 1)


s = input("Enter string: ")
print("You Entered :",s)
length = 0
for ch in s:
    length += 1

if palindrome(s, 0, length - 1):
    print(s,"is Palindrome")
else:
    print(s,"is Not Palindrome")
```
