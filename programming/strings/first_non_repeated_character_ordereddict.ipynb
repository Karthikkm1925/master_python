<strong style="font-size:30px;color:green">Find the first non-repeated character in a string using OrderedDict</strong>

<strong style="font-size:25px;color:green">Using OrderedDict frequency dictionary</strong>

```python
from collections import OrderedDict

s = "swiss"
print("You Entered :", s)

od = OrderedDict()
for ch in s:
    od[ch] = od.get(ch, 0) + 1

first_non_repeated = None
for ch, count in od.items():
    if count == 1:
        first_non_repeated = ch
        break

if first_non_repeated:
    print("First non-repeated character:", first_non_repeated)
else:
    print("No non-repeated character found.")
```

<strong style="font-size:25px;color:green">Using OrderedDict with character tracking</strong>

```python
from collections import OrderedDict

s = "swiss"
print("You Entered :", s)

counts = OrderedDict()

for ch in s:
    if ch in counts:
        counts[ch] += 1
    else:
        counts[ch] = 1

result = next((ch for ch, count in counts.items() if count == 1), None)

if result:
    print("First non-repeated character:", result)
else:
    print("No non-repeated character found.")
```

<strong style="font-size:25px;color:green">Using function with OrderedDict</strong>

```python
from collections import OrderedDict


def first_non_repeating_character(s):
    freq = OrderedDict()
    for ch in s:
        freq[ch] = freq.get(ch, 0) + 1

    for ch in freq:
        if freq[ch] == 1:
            return ch
    return None


s = "swiss"
print("You Entered :", s)

result = first_non_repeating_character(s)

if result:
    print("First non-repeated character:", result)
else:
    print("No non-repeated character found.")
```
