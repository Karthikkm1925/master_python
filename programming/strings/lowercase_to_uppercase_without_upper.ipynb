<strong style="font-size:30px;color:green">Convert lowercase characters to uppercase without using upper()</strong>

<strong style="font-size:25px;color:green">Using ASCII value manipulation (ord and chr)</strong>

```python
s = "hello world 123!"
print("You Entered :", s)

result = ""
for ch in s:
    if "a" <= ch <= "z":
        result += chr(ord(ch) - 32)
    else:
        result += ch

print("Result :", result)
```

<strong style="font-size:25px;color:green">Using dictionary mapping</strong>

```python
s = "hello world 123!"
print("You Entered :", s)

lower_to_upper = {
    "a": "A", "b": "B", "c": "C", "d": "D", "e": "E", "f": "F", "g": "G",
    "h": "H", "i": "I", "j": "J", "k": "K", "l": "L", "m": "M", "n": "N",
    "o": "O", "p": "P", "q": "Q", "r": "R", "s": "S", "t": "T", "u": "U",
    "v": "V", "w": "W", "x": "X", "y": "Y", "z": "Z"
}

result = "".join([lower_to_upper.get(ch, ch) for ch in s])

print("Result :", result)
```

<strong style="font-size:25px;color:green">Using str.maketrans() and translate()</strong>

```python
s = "hello world 123!"
print("You Entered :", s)

lower = "abcdefghijklmnopqrstuvwxyz"
upper = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
trans_table = str.maketrans(lower, upper)

result = s.translate(trans_table)

print("Result :", result)
```

<strong style="font-size:25px;color:green">Using bitwise AND operation</strong>

```python
s = "hello world 123!"
print("You Entered :", s)

result = ""
for ch in s:
    if "a" <= ch <= "z":
        result += chr(ord(ch) & ~32)
    else:
        result += ch

print("Result :", result)
```

<strong style="font-size:25px;color:green">Using recursion</strong>

```python
def to_upper(string):
    if not string:
        return ""
    ch = string[0]
    new_ch = chr(ord(ch) - 32) if "a" <= ch <= "z" else ch
    return new_ch + to_upper(string[1:])


s = "hello world 123!"
print("You Entered :", s)

result = to_upper(s)

print("Result :", result)
```
