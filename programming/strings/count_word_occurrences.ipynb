<strong style="font-size:30px;color:green">Count the occurrences of each word in a sentence</strong>

<strong style="font-size:25px;color:green">Using a dictionary</strong>

```python
sentence = "apple banana apple orange banana apple"
print("You Entered :", sentence)

words = []
word = ""

for ch in sentence:
    if ch != " ":
        word = word + ch
    else:
        if word != "":
            words.append(word)
            word = ""

if word != "":
    words.append(word)

count = {}

for word in words:
    if word in count:
        count[word] = count[word] + 1
    else:
        count[word] = 1

print(count)
```

<strong style="font-size:25px;color:green">Without dictionary</strong>

```python
sentence = "apple banana apple orange banana apple"
print("You Entered :", sentence)

words = []
word = ""

for ch in sentence:
    if ch != " ":
        word = word + ch
    else:
        if word != "":
            words.append(word)
            word = ""

if word != "":
    words.append(word)

visited = []

for i in range(len(words)):

    if words[i] in visited:
        continue

    count = 0

    for j in range(len(words)):
        if words[i] == words[j]:
            count = count + 1

    print(words[i], ":", count)

    visited.append(words[i])
```

<strong style="font-size:25px;color:green">Using nested loops — pure logic</strong>

```python
sentence = "apple banana apple orange banana apple"
print("You Entered :", sentence)

words = []
word = ""

for ch in sentence:
    if ch != " ":
        word = word + ch
    else:
        if word != "":
            words.append(word)
            word = ""

if word != "":
    words.append(word)

for i in range(len(words)):

    already_counted = False

    for k in range(i):
        if words[k] == words[i]:
            already_counted = True
            break

    if already_counted:
        continue

    count = 0

    for j in range(len(words)):
        if words[i] == words[j]:
            count = count + 1

    print(words[i], ":", count)
```

<strong style="font-size:25px;color:green">Using dictionary with manual word extraction</strong>

```python
sentence = "apple banana apple orange banana apple"
print("You Entered :", sentence)

count = {}
word = ""

for ch in sentence + " ":

    if ch != " ":
        word = word + ch

    else:

        if word != "":

            if word in count:
                count[word] = count[word] + 1
            else:
                count[word] = 1

            word = ""

print(count)
```

<strong style="font-size:25px;color:green">Case-insensitive version without .lower()</strong>

```python
sentence = "Apple apple APPLE Banana banana"
print("You Entered :", sentence)

count = {}
word = ""

for ch in sentence + " ":

    if ch != " ":

        if 'A' <= ch <= 'Z':
            ch = chr(ord(ch) + 32)

        word = word + ch

    else:

        if word != "":

            if word in count:
                count[word] = count[word] + 1
            else:
                count[word] = 1

            word = ""

print(count)
```
