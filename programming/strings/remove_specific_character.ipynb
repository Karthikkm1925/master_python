<strong style="font-size:30px;color:green">Remove all occurrences of a specific character from a string</strong>

<strong style="font-size:25px;color:green">Using replace() method</strong>

```python
s = "programming"
target = "m"
print("You Entered :", s)
print("Target Character :", target)

result = s.replace(target, "")

print("Result :", result)
```

<strong style="font-size:25px;color:green">Using a for loop and concatenation</strong>

```python
s = "programming"
target = "m"
print("You Entered :", s)
print("Target Character :", target)

result = ""
for ch in s:
    if ch != target:
        result += ch

print("Result :", result)
```

<strong style="font-size:25px;color:green">Using list comprehension and join()</strong>

```python
s = "programming"
target = "m"
print("You Entered :", s)
print("Target Character :", target)

result = "".join([ch for ch in s if ch != target])

print("Result :", result)
```

<strong style="font-size:25px;color:green">Using filter() and join()</strong>

```python
s = "programming"
target = "m"
print("You Entered :", s)
print("Target Character :", target)

result = "".join(filter(lambda ch: ch != target, s))

print("Result :", result)
```

<strong style="font-size:25px;color:green">Using recursion</strong>

```python
def remove_char(string, target):
    if not string:
        return ""
    if string[0] == target:
        return remove_char(string[1:], target)
    return string[0] + remove_char(string[1:], target)


s = "programming"
target = "m"
print("You Entered :", s)
print("Target Character :", target)

result = remove_char(s, target)

print("Result :", result)
```
