<strong style="font-size:30px;color:green">Remove all spaces from a string without using replace()</strong>

<strong style="font-size:25px;color:green">Using split() and join()</strong>

```python
s = "Python Data Analysis"
result = "".join(s.split())
print("You Entered :", s)
print("Result :", result)
```

<strong style="font-size:25px;color:green">Using a for loop and concatenation</strong>

```python
s = "Python Data Analysis"
result = ""
for ch in s:
    if ch != " ":
        result += ch

print("You Entered :", s)
print("Result :", result)
```

<strong style="font-size:25px;color:green">Using list comprehension and join()</strong>

```python
s = "Python Data Analysis"
result = "".join([ch for ch in s if ch != " "])
print("You Entered :", s)
print("Result :", result)
```

<strong style="font-size:25px;color:green">Using filter() and join()</strong>

```python
s = "Python Data Analysis"
result = "".join(filter(lambda ch: ch != " ", s))
print("You Entered :", s)
print("Result :", result)
```

<strong style="font-size:25px;color:green">Using recursion</strong>

```python
def remove_spaces(string):
    if not string:
        return ""
    if string[0] == " ":
        return remove_spaces(string[1:])
    return string[0] + remove_spaces(string[1:])


s = "Python Data Analysis"
result = remove_spaces(s)

print("You Entered :", s)
print("Result :", result)
```
