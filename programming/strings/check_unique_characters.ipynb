<strong style="font-size:30px;color:green">Check if a string contains only unique characters</strong>

<strong style="font-size:25px;color:green">Using set() and len() comparison</strong>

```python
s = "python"
print("You Entered :", s)

is_unique = len(s) == len(set(s))

if is_unique:
    print("The string contains only unique characters.")
else:
    print("The string contains duplicate characters.")
```

<strong style="font-size:25px;color:green">Using a set with early exit</strong>

```python
s = "python"
print("You Entered :", s)

seen = set()
is_unique = True

for ch in s:
    if ch in seen:
        is_unique = False
        break
    seen.add(ch)

if is_unique:
    print("The string contains only unique characters.")
else:
    print("The string contains duplicate characters.")
```

<strong style="font-size:25px;color:green">Without built-in data structures</strong>

```python
s = "python"
print("You Entered :", s)

is_unique = True

for i in range(len(s)):
    for j in range(i + 1, len(s)):
        if s[i] == s[j]:
            is_unique = False
            break
    if not is_unique:
        break

if is_unique:
    print("The string contains only unique characters.")
else:
    print("The string contains duplicate characters.")
```

<strong style="font-size:25px;color:green">Using sorting</strong>

```python
s = "python"
print("You Entered :", s)

sorted_s = sorted(s)
is_unique = True

for i in range(len(sorted_s) - 1):
    if sorted_s[i] == sorted_s[i + 1]:
        is_unique = False
        break

if is_unique:
    print("The string contains only unique characters.")
else:
    print("The string contains duplicate characters.")
```

<strong style="font-size:25px;color:green">Using string count() method</strong>

```python
s = "python"
print("You Entered :", s)

is_unique = True

for ch in s:
    if s.count(ch) > 1:
        is_unique = False
        break

if is_unique:
    print("The string contains only unique characters.")
else:
    print("The string contains duplicate characters.")
```
