<strong style="font-size:30px;color:green">Convert uppercase characters to lowercase without using lower()</strong>

<strong style="font-size:25px;color:green">Using ASCII value manipulation (ord and chr)</strong>

```python
s = "HELLO WORLD 123!"
print("You Entered :", s)

result = ""
for ch in s:
    if "A" <= ch <= "Z":
        result += chr(ord(ch) + 32)
    else:
        result += ch

print("Result :", result)
```

<strong style="font-size:25px;color:green">Using dictionary mapping</strong>

```python
s = "HELLO WORLD 123!"
print("You Entered :", s)

upper_to_lower = {
    "A": "a", "B": "b", "C": "c", "D": "d", "E": "e", "F": "f", "G": "g",
    "H": "h", "I": "i", "J": "j", "K": "k", "L": "l", "M": "m", "N": "n",
    "O": "o", "P": "p", "Q": "q", "R": "r", "S": "s", "T": "t", "U": "u",
    "V": "v", "W": "w", "X": "x", "Y": "y", "Z": "z"
}

result = "".join([upper_to_lower.get(ch, ch) for ch in s])

print("Result :", result)
```

<strong style="font-size:25px;color:green">Using str.maketrans() and translate()</strong>

```python
s = "HELLO WORLD 123!"
print("You Entered :", s)

upper = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
lower = "abcdefghijklmnopqrstuvwxyz"
trans_table = str.maketrans(upper, lower)

result = s.translate(trans_table)

print("Result :", result)
```

<strong style="font-size:25px;color:green">Using bitwise OR operation</strong>

```python
s = "HELLO WORLD 123!"
print("You Entered :", s)

result = ""
for ch in s:
    if "A" <= ch <= "Z":
        result += chr(ord(ch) | 32)
    else:
        result += ch

print("Result :", result)
```

<strong style="font-size:25px;color:green">Using recursion</strong>

```python
def to_lower(string):
    if not string:
        return ""
    ch = string[0]
    new_ch = chr(ord(ch) + 32) if "A" <= ch <= "Z" else ch
    return new_ch + to_lower(string[1:])


s = "HELLO WORLD 123!"
print("You Entered :", s)

result = to_lower(s)

print("Result :", result)
```
