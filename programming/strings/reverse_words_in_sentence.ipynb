<strong style="font-size:30px;color:green">Reverse the order of words in a sentence</strong>

<strong style="font-size:25px;color:green">Using a loop and building words</strong>

```python
s = "I love Python programming"
print("You Entered :", s)

words = []
word = ""

for ch in s:
    if ch == " ":
        words.append(word)
        word = ""
    else:
        word += ch

words.append(word)

result = ""

i = len(words) - 1

while i >= 0:
    result += words[i]

    if i != 0:
        result += " "

    i -= 1

print(result)
```

<strong style="font-size:25px;color:green">Without storing words in a list</strong>

```python
s = "Coding in Python is fun"
print("You Entered :", s)

end = len(s) - 1
result = ""

while end >= 0:

    while end >= 0 and s[end] == " ":
        end -= 1

    start = end

    while start >= 0 and s[start] != " ":
        start -= 1

    i = start + 1

    while i <= end:
        result += s[i]
        i += 1

    if start > 0:
        result += " "

    end = start - 1

print(result)
```

<strong style="font-size:25px;color:green">Using a list as a stack</strong>

```python
s = "Artificial Intelligence will change the world"
print("You Entered :", s)

words = []
word = ""

for ch in s:
    if ch == " ":
        words.append(word)
        word = ""
    else:
        word += ch

words.append(word)

result = ""

while len(words) > 0:
    result += words[-1]
    words.pop()

    if len(words) > 0:
        result += " "

print(result)
```

<strong style="font-size:25px;color:green">Reverse the entire string, then reverse each word</strong>

```python
s = "Reverse the order of words"
print("You Entered :", s)

a = []

for ch in s:
    a.append(ch)

left = 0
right = len(a) - 1

while left < right:
    temp = a[left]
    a[left] = a[right]
    a[right] = temp

    left += 1
    right -= 1

start = 0
i = 0

while i <= len(a):

    if i == len(a) or a[i] == " ":

        left = start
        right = i - 1

        while left < right:
            temp = a[left]
            a[left] = a[right]
            a[right] = temp

            left += 1
            right -= 1

        start = i + 1

    i += 1

result = ""

for ch in a:
    result += ch

print(result)
```

<strong style="font-size:25px;color:green">Recursive solution</strong>

```python
s = "Practice makes a person perfect"
print("You Entered :", s)


def reverse_words(s, i, word):

    if i < 0:
        print(word, end="")
        return

    if s[i] == " ":
        print(word, end=" ")

        reverse_words(s, i - 1, "")

    else:
        reverse_words(s, i - 1, s[i] + word)


reverse_words(s, len(s) - 1, "")
```

<strong style="font-size:25px;color:green">Using recursion with returned result</strong>

```python
s = "Keep pushing your limits daily"
print("You Entered :", s)


def reverse_words(s, i, result):

    if i < 0:
        return result

    word = ""

    while i >= 0 and s[i] != " ":
        word = s[i] + word
        i -= 1

    if result == "":
        result = word
    else:
        result = result + " " + word

    while i >= 0 and s[i] == " ":
        i -= 1

    return reverse_words(s, i, result)


print(reverse_words(s, len(s) - 1, ""))
```

<strong style="font-size:25px;color:green">Using for loop from right to left</strong>

```python
s = "Never give up on learning"
print("You Entered :", s)

result = ""
word = ""

i = len(s) - 1

while i >= 0:

    if s[i] != " ":
        word = s[i] + word

    else:
        if word != "":
            if result == "":
                result = word
            else:
                result = result + " " + word

            word = ""

    i -= 1

if word != "":
    if result == "":
        result = word
    else:
        result = result + " " + word

print(result)
```
