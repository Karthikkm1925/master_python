<strong style="font-size:30px;color:green">Find the longest and shortest words in a sentence</strong>

<strong style="font-size:25px;color:green">Using max() and min() with key=len</strong>

```python
sentence = "Python is a great programming language"
print("You Entered :", sentence)

words = sentence.split()
longest_word = max(words, key=len)
shortest_word = min(words, key=len)

print("Longest Word :", longest_word)
print("Shortest Word :", shortest_word)
```

<strong style="font-size:25px;color:green">Single pass manual tracking of longest and shortest</strong>

```python
sentence = "Python is a great programming language"
print("You Entered :", sentence)

words = sentence.split()
longest_word = words[0] if words else ""
shortest_word = words[0] if words else ""

for word in words[1:]:
    if len(word) > len(longest_word):
        longest_word = word
    if len(word) < len(shortest_word):
        shortest_word = word

print("Longest Word :", longest_word)
print("Shortest Word :", shortest_word)
```

<strong style="font-size:25px;color:green">Using sorted() with key=len</strong>

```python
sentence = "Python is a great programming language"
print("You Entered :", sentence)

words = sentence.split()
sorted_words = sorted(words, key=len)

shortest_word = sorted_words[0] if sorted_words else ""
longest_word = sorted_words[-1] if sorted_words else ""

print("Longest Word :", longest_word)
print("Shortest Word :", shortest_word)
```

<strong style="font-size:25px;color:green">Without using split() - manual word extraction</strong>

```python
sentence = "Python is a great programming language"
print("You Entered :", sentence)

words = []
current_word = ""

for ch in sentence:
    if ch != " ":
        current_word += ch
    else:
        if current_word != "":
            words.append(current_word)
            current_word = ""

if current_word != "":
    words.append(current_word)

longest_word = words[0] if words else ""
shortest_word = words[0] if words else ""

for w in words[1:]:
    if len(w) > len(longest_word):
        longest_word = w
    if len(w) < len(shortest_word):
        shortest_word = w

print("Longest Word :", longest_word)
print("Shortest Word :", shortest_word)
```

<strong style="font-size:25px;color:green">Without using built-in len() or split()</strong>

```python
sentence = "Python is a great programming language"
print("You Entered :", sentence)

words = []
current_word = ""

for ch in sentence:
    if ch != " ":
        current_word += ch
    else:
        if current_word != "":
            words.append(current_word)
            current_word = ""

if current_word != "":
    words.append(current_word)

longest_word = ""
shortest_word = ""
max_len = 0
min_len = float("inf")

for w in words:
    w_len = 0
    for _ in w:
        w_len += 1

    if w_len > max_len:
        max_len = w_len
        longest_word = w

    if w_len < min_len:
        min_len = w_len
        shortest_word = w

print("Longest Word :", longest_word)
print("Shortest Word :", shortest_word)
```
