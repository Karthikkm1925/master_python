<strong style="font-size:30px;color:green">Find the most frequent word in a sentence</strong>

<strong style="font-size:25px;color:green">Using collections.Counter and most_common()</strong>

```python
from collections import Counter

sentence = "data science uses python and python is popular because python is versatile"
print("You Entered :", sentence)

words = sentence.split()
counts = Counter(words)
most_frequent, count = counts.most_common(1)[0]

print("Most Frequent Word :", most_frequent)
print("Frequency :", count)
```

<strong style="font-size:25px;color:green">Using a frequency dictionary</strong>

```python
sentence = "data science uses python and python is popular because python is versatile"
print("You Entered :", sentence)

words = sentence.split()
freq = {}

for word in words:
    freq[word] = freq.get(word, 0) + 1

most_frequent = ""
highest_count = 0

for word, count in freq.items():
    if count > highest_count:
        highest_count = count
        most_frequent = word

print("Most Frequent Word :", most_frequent)
print("Frequency :", highest_count)
```

<strong style="font-size:25px;color:green">Using max() with key=words.count</strong>

```python
sentence = "data science uses python and python is popular because python is versatile"
print("You Entered :", sentence)

words = sentence.split()
most_frequent = max(set(words), key=words.count)
count = words.count(most_frequent)

print("Most Frequent Word :", most_frequent)
print("Frequency :", count)
```

<strong style="font-size:25px;color:green">Without using split() - manual word extraction and dictionary</strong>

```python
sentence = "data science uses python and python is popular because python is versatile"
print("You Entered :", sentence)

words = []
current_word = ""

for ch in sentence:
    if ch != " ":
        current_word += ch
    else:
        if current_word != "":
            words.append(current_word)
            current_word = ""

if current_word != "":
    words.append(current_word)

freq = {}
for w in words:
    freq[w] = freq.get(w, 0) + 1

most_frequent = ""
highest_count = 0

for w, c in freq.items():
    if c > highest_count:
        highest_count = c
        most_frequent = w

print("Most Frequent Word :", most_frequent)
print("Frequency :", highest_count)
```

<strong style="font-size:25px;color:green">Without using built-in functions or dictionary</strong>

```python
sentence = "data science uses python and python is popular because python is versatile"
print("You Entered :", sentence)

words = []
current_word = ""

for ch in sentence:
    if ch != " ":
        current_word += ch
    else:
        if current_word != "":
            words.append(current_word)
            current_word = ""

if current_word != "":
    words.append(current_word)

unique_words = []
counts = []

for w in words:
    found = False
    for i in range(len(unique_words)):
        if unique_words[i] == w:
            counts[i] += 1
            found = True
            break
    if not found:
        unique_words.append(w)
        counts.append(1)

highest_count = 0
most_frequent = ""

for i in range(len(counts)):
    if counts[i] > highest_count:
        highest_count = counts[i]
        most_frequent = unique_words[i]

print("Most Frequent Word :", most_frequent)
print("Frequency :", highest_count)
```
